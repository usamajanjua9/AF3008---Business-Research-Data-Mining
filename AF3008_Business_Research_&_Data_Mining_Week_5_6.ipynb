# **1. Introduction to Business Data**

---

## **1.1 Meaning and Importance of Data in Business Decision-Making**

### 🔎 What is Business Data?

Business data refers to **information collected by organizations** that helps them:

* Understand customers
* Monitor performance
* Improve operations
* Make strategic decisions

---

### 🧠 Simple Explanation

> Business data is any information that helps a company make better decisions.

---

### 📊 Examples of Business Data

* Sales revenue
* Number of transactions
* Customer complaints
* Website visits
* Payment records
* Customer demographics

---

### 🏦 FinTech Example

PaySecure collects:

* Transaction amount

* Fraud detection alerts

## **1.2 Role of Data in Analytics and FinTech**

FinTech heavily depends on data for:

* Fraud detection
* Credit scoring
* Risk analysis
* Customer segmentation
* Personalized recommendations

---

### 📌 Example

Credit scoring systems analyze:

* Transaction history
* Income patterns
* Repayment behavior

To predict default risk.

---

### 📈 In Analytics

Data is used to:

* Identify trends
* Predict outcomes
* Optimize pricing
* Improve marketing campaigns

---

## **1.3 Data-Driven vs Intuition-Based Decisions**

---

### 🔎 Data-Driven Decisions

Decisions based on:

* Evidence
* Statistical analysis
* Measurable results

Example:

Data shows 70% users abandon wallet after first transaction → improve onboarding.

---

### 🔎 Intuition-Based Decisions

Decisions based on:

* Manager’s experience
* Personal opinion
* Assumptions

Example:

“I think students don’t use wallet because they don’t like the color.”

---

### 📌 Comparison

| Data-Driven    | Intuition-Based  |
| -------------- | ---------------- |
| Objective      | Subjective       |
| Evidence-based | Assumption-based |
| Measurable     | Opinion-driven   |
| Lower risk     | Higher risk      |

Modern business prefers data-driven decision-making.

# **3. Structured and Unstructured Data**



---

# **3.1 Structured Data**

---

## 🔎 Definition of Structured Data

Structured data is data that is:

* Organized in a fixed format
* Stored in rows and columns
* Easily searchable and analyzable
* Stored in relational databases

It follows a predefined schema (structure).

---

### 🧠 Simple Explanation

> Structured data is neatly organized like an Excel sheet.

---

## 📊 What Structured Data Looks Like

Example: Digital Wallet Transactions

| User_ID | Age | Monthly_Income | Transactions | Avg_Amount |
| ------- | --- | -------------- | ------------ | ---------- |
| U001    | 21  | 30000          | 15           | 1200       |
| U002    | 23  | 45000          | 7            | 2100       |
| U003    | 20  | 25000          | 20           | 950        |

This data:

* Has fixed columns
* Each column has a defined data type
* Can be easily imported into Python, Excel, SPSS, etc.

---

## 🏦 Examples of Structured Business Data

* Sales records
* Bank transaction logs
* Customer demographics
* Inventory lists
* Financial statements
* Credit score tables
* Employee payroll data

---

## 📌 Where Structured Data is Stored

* SQL databases
* Excel files
* CSV files
* ERP systems
* Banking systems

---

# **3.2 Unstructured Data**

---

## 🔎 Definition of Unstructured Data

Unstructured data does not have a predefined format or organized schema.

It cannot be easily stored in rows and columns.

---

### 🧠 Simple Explanation

> Unstructured data is messy data that does not fit neatly into a table.

---

## 📊 Examples of Unstructured Data

* Customer reviews
* Social media posts
* Emails
* Chat messages
* Images
* Videos
* Audio recordings
* Call center transcripts

---

### 🏦 FinTech Example

PaySecure collects:

Customer feedback:

“I am worried about fraud. I don’t trust digital wallets.”

This is text data — not organized into numbers.

---

## 📌 What Unstructured Data Looks Like

Example: Customer Review Dataset

| Review_Text                    |
| ------------------------------ |
| “App crashes frequently.”      |
| “Very secure and easy to use.” |
| “Customer support is slow.”    |

No numeric structure — requires processing before analysis.

---

# **3.3 Semi-Structured Data (Brief Introduction)**

---

## 🔎 Definition

Semi-structured data is partially organized but does not fit traditional relational databases.

It contains tags or markers to separate elements.

---

### 📌 Examples

* JSON files
* XML files
* API responses
* NoSQL databases

---

### 📊 Example (JSON format)

```
{
  "User_ID": "U001",
  "Transactions": 15,
  "Feedback": "App is secure"
}
```

It has structure but is not in strict table format.

---

# **3.4 Challenges in Handling Unstructured Data**

Unstructured data creates several challenges:

---

### 1️⃣ Requires Preprocessing

Text must be cleaned before analysis.

Example:

* Remove stop words
* Convert to lowercase
* Remove punctuation

---

### 2️⃣ Requires Advanced Techniques

* Natural Language Processing (NLP)
* Image recognition
* Sentiment analysis
* Machine learning

---

### 3️⃣ Storage Challenges

Large storage required for:

* Videos
* Images
* Audio

---

### 4️⃣ Difficult to Quantify

You cannot directly run regression on:

“I like this app.”

It must first be converted into numerical format.

---

# **3.5 Importance of Structured Data in Statistical Analysis**

Structured data is essential because:

* It supports statistical analysis
* It works with Python (Pandas, NumPy)
* It enables regression, correlation, hypothesis testing
* It is easier to visualize

---

## 📌 Example

To test:

H₁: Trust increases wallet usage.

We need structured data:

| Trust_Score | Usage_Frequency |
| ----------- | --------------- |
| 4           | 12              |
| 2           | 3               |

Only structured numeric data allows:

* Correlation
* Regression
* Hypothesis testing

---

## 📊 Why Data Mining Prefers Structured Data

Data mining algorithms require:

* Numerical input
* Organized variables
* Clean datasets

That is why unstructured data must first be transformed.

---

# 📊 Complete Comparison Table

| Feature                 | Structured Data  | Unstructured Data | Semi-Structured Data |
| ----------------------- | ---------------- | ----------------- | -------------------- |
| Format                  | Fixed schema     | No schema         | Partial structure    |
| Storage                 | Relational DB    | Text files, media | JSON, XML            |
| Analysis                | Easy             | Difficult         | Moderate             |
| Examples                | Transaction logs | Reviews, images   | API responses        |
| Suitable for statistics | Yes              | After processing  | Sometimes            |

---

# 🎯 Key Takeaways

* Structured data = organized and easy to analyze.
* Unstructured data = messy and requires processing.
* Semi-structured data = partially organized.
* Statistical testing requires structured data.
* FinTech systems generate massive structured transaction data.

# **4.1 Financial Data**

---

## 🔎 What is Financial Data?

Financial data represents a company’s **financial health and performance**.

It answers:

* Is the company profitable?
* How much cash is available?
* What are the liabilities?

---

## 📊 Main Financial Statements

---

### 1️⃣ Income Statement

Shows:

* Revenue
* Expenses
* Profit or loss

Example:

| Year | Revenue | Expenses | Net Profit |
| ---- | ------- | -------- | ---------- |
| 2024 | 50M     | 35M      | 15M        |

---

### 2️⃣ Balance Sheet

Shows:

* Assets
* Liabilities
* Equity

Basic Formula:

Assets = Liabilities + Equity

Example:

| Assets | Liabilities | Equity |
| ------ | ----------- | ------ |
| 100M   | 60M         | 40M    |

---

### 3️⃣ Cash Flow Statement

Shows:

* Cash inflows
* Cash outflows

Example:

| Operating Cash | Investing Cash | Financing Cash |
| -------------- | -------------- | -------------- |
| +10M           | -5M            | +2M            |

---

### 🏦 FinTech Example

A digital wallet company tracks:

* Transaction revenue
* Fraud losses
* Operating expenses
* Cash reserves

## 📌 Why Financial Data Matters

* Profitability analysis
* Risk assessment
* Investment decisions
* Strategic expansion planning

# **Week 6 – Data Collection Methods**

## **1. Introduction to Data Collection**

---

# **1. Introduction to Data Collection**

Data collection is the stage where research moves from theory to evidence.

After defining:

* Research problem
* Research questions
* Research design
* Hypotheses

We must gather data to test our assumptions.

Without proper data collection, research conclusions are meaningless.

---

# **1.1 Purpose of Data Collection in Research**

---

## 🔎 What is Data Collection?

Data collection is the systematic process of gathering information to answer research questions and test hypotheses.

---

## 🧠 Simple Explanation

> Data collection means collecting evidence to support or reject your research idea.

---

## 🎯 Main Purposes of Data Collection

### 1️⃣ To Answer Research Questions

Example:

Research Question:
Why are students not adopting digital wallets?

Data needed:

* Survey responses
* Usage statistics
* Interview feedback

---

### 2️⃣ To Test Hypotheses

Hypothesis:
Perceived security positively affects wallet usage.

Data required:

* Trust score
* Transaction frequency

Without data → hypothesis cannot be tested.

---

### 3️⃣ To Support Managerial Decisions

Businesses use research data to:

* Launch new products
* Improve service quality
* Reduce fraud
* Increase customer retention

---

### 🏦 Business Example

A FinTech company wants to know:

Should we introduce cashback incentives?

They collect:

* User feedback
* Transaction patterns
* Experimental data

Data guides decision.

---

# **1.2 Linking Research Design to Data Collection Method**

Data collection must match research design.

Different research designs require different data collection approaches.

---

## 🔵 Exploratory Research

Purpose:
Explore unclear problems.

Suitable methods:

* Interviews
* Focus groups
* Open-ended surveys
* Observations

Example:
Why do rural users avoid digital banking?

---

## 🟢 Descriptive Research

Purpose:
Describe characteristics.

Suitable methods:

* Structured surveys
* Secondary datasets
* Observational counts

Example:
What percentage of students use digital wallets?

---

## 🔴 Causal Research

Purpose:
Test cause-effect relationships.

Suitable methods:

* Experiments
* A/B testing
* Field trials

Example:
Does cashback increase usage?

---

## 📊 Alignment Table

| Research Design | Data Collection Method |
| --------------- | ---------------------- |
| Exploratory     | Interviews             |
| Descriptive     | Surveys                |
| Causal          | Experiments            |

If method does not match design → research becomes inconsistent.

---

# **1.3 Reliability and Validity Considerations**

Collecting data is not enough.

Data must be:

* Reliable
* Valid

Otherwise results are misleading.

---

# **Reliability**

---

## 🔎 What is Reliability?

Reliability refers to the consistency of measurement.

If the same measurement is repeated, results should be similar.

---

## 🧠 Example

If a student rates trust level today as 4/5
And next week also rates 4/5

Measurement is reliable.

---

## 🔴 Problem Example

If responses change drastically without reason, instrument is unreliable.

---

## 📌 Types of Reliability (Basic)

* Test-retest reliability
* Internal consistency
* Inter-rater reliability

---

# **Validity**

---

## 🔎 What is Validity?

Validity refers to accuracy.

Are we measuring what we intend to measure?

---

## 🧠 Example

If we want to measure “Trust” but only ask about “App color design” → not valid.

---

## 📌 Types of Validity (Intro Level)

* Content validity
* Construct validity
* Internal validity
* External validity

---

# 🔵 Reliability vs Validity (Simple Comparison)

| Concept     | Meaning     |
| ----------- | ----------- |
| Reliability | Consistency |
| Validity    | Accuracy    |

---

## 🎯 Important Rule

Measurement can be:

Reliable but not valid
Valid but not reliable
Reliable and valid (ideal)

---

### 🎯 Target Analogy

* Hitting same wrong spot repeatedly → Reliable but not valid
* Hitting random spots around center → Valid but not reliable
* Hitting center consistently → Reliable and valid

---

# 📌 Why This Matters in Business Research

If survey is:

* Poorly worded → validity issue
* Inconsistent → reliability issue

Then:

* Statistical results become meaningless
* Managerial decisions become flawed

---

# 📊 Conceptual Flow

Research Problem → Research Design → Data Collection → Reliability & Validity → Analysis → Decision

---

# 🎯 Key Takeaways

* Data collection converts theory into measurable evidence.
* Method must align with research design.
* Reliability ensures consistency.
* Validity ensures accuracy.
* Poor data collection leads to incorrect conclusions.

# **2. Surveys and Questionnaires**

Surveys are the most widely used data collection method in business research, marketing research, and FinTech studies.

They are especially useful in **descriptive research**.

---

# **2.1 Meaning of Surveys**

---

## 🔎 What is a Survey?

A survey is a structured method of collecting data from respondents using a standardized set of questions.

All participants answer the same questions in the same format.

---

## 🧠 Simple Explanation

> A survey collects structured responses from many people to identify patterns and trends.

---

## 🏦 Business Example

Research Question:
Does perceived security affect digital wallet usage?

Survey collects:

* Trust score (1–5 scale)
* Monthly transaction count
* Age, gender

The collected data is then analyzed statistically.

---

# **2.2 Structure of a Questionnaire**

A well-designed questionnaire follows a logical structure.

---

## 📌 Typical Structure

### 1️⃣ Introduction

* Purpose of research
* Confidentiality assurance
* Consent statement

Example:
“This survey is for academic research. Your responses will remain confidential.”

---

### 2️⃣ Screening Questions

Ensures respondent is relevant.

Example:
Do you use digital wallets?
☐ Yes
☐ No

---

### 3️⃣ Demographic Section

* Age
* Gender
* Education
* Income

---

### 4️⃣ Main Research Variables

Questions measuring:

* Trust
* Satisfaction
* Perceived usefulness
* Behavioral intention

---

### 5️⃣ Closing Statement

“Thank you for your participation.”

---

# **2.3 Types of Survey Questions**

Different question types produce different types of data.

---

## 1️⃣ Closed-Ended Questions

Provide fixed response options.

Example:
Do you use digital wallets?
☐ Yes
☐ No

Produces nominal data.

---

## 2️⃣ Likert Scale Questions

Measure level of agreement or intensity.

Example:

“I believe PaySecure is secure.”

1 – Strongly Disagree
2 – Disagree
3 – Neutral
4 – Agree
5 – Strongly Agree

Produces ordinal data.

Most common in business research.

---

## 3️⃣ Multiple Choice Questions

Respondent selects one or more options.

Example:
Which wallet do you use most?
☐ PaySecure
☐ JazzCash
☐ EasyPaisa
☐ Other

Produces nominal data.

---

## 4️⃣ Open-Ended Questions

Respondent writes answer in own words.

Example:
Why do you prefer digital wallets?

Produces qualitative data.

---

# **2.4 Designing Effective Questionnaires**

Poor questionnaire design leads to invalid research.

---

## 📌 Guidelines for Good Design

### ✔ Keep Questions Clear

Avoid complex or technical language.

Bad:
“Do you perceive fintech adoption to be technologically secure?”

Better:
“Do you think digital wallets are secure?”

---

### ✔ Avoid Leading Questions

Bad:
“Don’t you think PaySecure is very secure?”

Good:
“How would you rate the security of PaySecure?”

---

### ✔ Avoid Double-Barreled Questions

Bad:
“Is PaySecure fast and secure?”

This measures two things.

Better:
Separate into two questions.

---

### ✔ Use Logical Flow

* Start with easy questions
* Place sensitive questions later
* Group similar topics

---

### ✔ Keep Survey Length Reasonable

Long surveys reduce response quality.

# **3. Interviews and Observational Studies**

Qualitative methods help researchers understand **why** and **how** something happens, not just **how much**.

They are especially useful in:

* Exploratory research
* Early-stage business problems
* Understanding customer perceptions
* FinTech adoption behavior

---

# **3.1 Interviews**

---

## 🔎 What is an Interview?

An interview is a structured or flexible conversation used to collect in-depth information from respondents.

---

### 🧠 Simple Explanation

> Interviews allow researchers to explore thoughts, feelings, and motivations behind behavior.

---

## **Types of Interviews**

---

### 1️⃣ Structured Interviews

* Predefined questions
* Same questions asked to every participant
* Little flexibility

📌 Best for:

* Comparisons across respondents
* Semi-quantitative analysis

Example (FinTech):

“On a scale of 1–5, how secure do you feel using digital wallets?”

---

### 2️⃣ Semi-Structured Interviews

* Prepared questions
* Flexible follow-up questions
* Allows deeper exploration

📌 Most common in business research.

Example:

“Why do you feel hesitant using digital wallets?”
(Interviewer probes deeper based on answer)

---

### 3️⃣ Unstructured Interviews

* No predefined questions
* Conversational format
* Open exploration

📌 Used in exploratory research.

Example:

“Tell me about your experience using digital payment apps.”

---

## 📊 Comparison of Interview Types

| Type            | Flexibility | Structure | Depth     |
| --------------- | ----------- | --------- | --------- |
| Structured      | Low         | High      | Moderate  |
| Semi-Structured | Medium      | Medium    | High      |
| Unstructured    | High        | Low       | Very High |

# **3.2 Focus Group Discussions**

---

## 🔎 What is a Focus Group?

A focus group is a moderated discussion involving 6–10 participants discussing a specific topic.

---

### 🧠 Purpose

* Explore collective opinions
* Generate new insights
* Understand group dynamics

---

### 🏦 FinTech Example

Group discussion among students:

“Why do you prefer EasyPaisa over other wallets?”

Participants may influence each other’s responses, revealing shared perceptions.

---

### 📌 Advantages

* Rich insights
* Multiple perspectives
* Efficient for exploratory research

---

### ❌ Limitations

* Dominant participants may influence discussion
* Harder to analyze
* Less privacy

# **4. Experiments and Field Studies**

Experiments are primarily used in **causal research design** to test cause-and-effect relationships.

They answer:

> Does X cause Y?

---

# **4.1 Laboratory Experiments**

---

## 🔎 What is a Laboratory Experiment?

A laboratory experiment is conducted in a **controlled environment** where the researcher manipulates one variable and observes its effect on another.

---

### 🧠 Simple Explanation

> Lab experiments control everything except the variable being tested.

---

### 📌 Key Features

* High control
* Controlled environment
* Random assignment
* Clear manipulation of independent variable

---

### 🏦 Example (FinTech)

Research Question:
Does adding a security badge increase trust?

Design:

* Group A sees app with security badge
* Group B sees app without badge
* Measure trust score

This is a laboratory-style controlled experiment.

---

## ✅ Advantages

* High internal validity
* Clear cause-effect relationship
* Controlled variables

---

## ❌ Limitations

* Artificial environment
* May not reflect real-world behavior

---

# **4.2 Field Experiments**

---

## 🔎 What is a Field Experiment?

A field experiment is conducted in a **real-world environment**, but still includes manipulation and control.

---

### 🧠 Simple Explanation

> Field experiments test changes in real business settings.

---

### 🏦 Example

PaySecure introduces:

* New cashback offer for 50% of users
* Other 50% receive no offer

Compare usage behavior.

This is a field experiment.

---

## ✅ Advantages

* High external validity
* Real-world behavior
* Practical business insights

---

## ❌ Limitations

* Less control
* Harder to manage variables
* Possible ethical concerns

---

# **4.3 A/B Testing in Business**

---

## 🔎 What is A/B Testing?

A/B testing is a simplified experimental method widely used in digital business.

Two versions are compared:

* Version A (control)
* Version B (treatment)

---

### 🏦 Example

Test:

* Blue “Pay Now” button (A)
* Green “Pay Now” button (B)

Measure:

* Click rate
* Transaction completion rate

---

## 📊 A/B Testing Steps

1. Define hypothesis
2. Randomly divide users
3. Implement two versions
4. Measure performance
5. Compare results statistically

---

## 📌 Why A/B Testing is Important

* Data-driven decision-making
* Low risk testing
* Quick feedback

Used by:

* Banks
* E-commerce
* FinTech apps
* Digital platforms

---

# **4.4 Control and Treatment Groups**

---

## 🔎 Control Group

Receives no intervention.

Acts as baseline.

---

## 🔎 Treatment Group

Receives experimental change.

---

### 🏦 Example

| Group     | Feature     | Usage           |
| --------- | ----------- | --------------- |
| Control   | No cashback | 10 transactions |
| Treatment | Cashback    | 18 transactions |

If difference is statistically significant → intervention works.

---

# **4.5 Measuring Intervention Impact**

To measure impact:

1. Define dependent variable
2. Measure before and after
3. Compare control vs treatment
4. Use statistical tests

---

### 📊 Basic Formula (Conceptual)

Impact = Treatment Outcome – Control Outcome

---

### Example

Average usage (Treatment) = 18
Average usage (Control) = 10

Impact = 8 additional transactions

---

Statistical testing determines if this difference is significant.

---

# **4.6 Internal vs External Validity**

---

## 🔎 Internal Validity

Refers to:

> Whether the experiment correctly shows cause-effect.

High internal validity means:

* No confounding variables
* Proper randomization
* Controlled environment

Lab experiments usually have high internal validity.

---

## 🔎 External Validity

Refers to:

> Whether results can be generalized to real-world settings.

Field experiments usually have higher external validity.

---

## 📊 Comparison Table

| Type       | Internal Validity | External Validity |
| ---------- | ----------------- | ----------------- |
| Laboratory | High              | Low               |
| Field      | Moderate          | High              |

---

# 🎯 Key Takeaways

* Experiments test cause-and-effect relationships.
* Laboratory experiments maximize control.
* Field experiments maximize realism.
* A/B testing is widely used in digital business.
* Control and treatment groups are essential.
* Internal validity ensures correctness; external validity ensures generalization.

# **5. Ethical Considerations in Business Data Collection**

Ethics in research ensures that data collection is conducted responsibly, legally, and transparently.

Especially in **business, finance, and FinTech**, ethical compliance is critical because sensitive financial and personal data is involved.

---

# **5.1 Importance of Research Ethics**

### 🔎 What are Research Ethics?

Research ethics are principles that guide responsible behavior during data collection and analysis.

---

### 🧠 Simple Explanation

> Ethics ensure that research does not harm participants and does not misuse data.

---

### 📌 Why Ethics Matter in Business Research

* Protects participant rights
* Ensures data privacy
* Maintains research credibility
* Prevents legal issues
* Preserves institutional reputation

---

# **5.2 Informed Consent**

---

## 🔎 What is Informed Consent?

Participants must be clearly informed about:

* Purpose of research
* How data will be used
* Their rights
* Option to withdraw

Before participating.

---

### 📌 Example (Survey Consent Statement)

“This survey is for academic research. Your responses will remain confidential. You may withdraw at any time.”

---

### 🏦 FinTech Context

If collecting financial transaction data, participants must know:

* What data is collected
* How it will be stored
* Whether it will be anonymized

---

# **5.3 Confidentiality and Privacy**

---

## 🔎 Confidentiality

Ensuring participant identity is not disclosed.

Example:

Instead of storing name → use Respondent ID.

---

## 🔎 Privacy

Respecting personal and sensitive information.

Especially important in:

* Financial data
* Income levels
* Transaction history
* Credit scores

---

### 📌 Best Practices

* Anonymize datasets
* Encrypt financial data
* Limit data access
* Remove identifying details

---

# **5.4 Data Protection Regulations**

Business research must comply with data protection laws.

---

### 🌍 Examples of Regulations

* GDPR (Europe)
* CCPA (California)
* Local financial regulations
* Central bank compliance rules

---

### 🏦 Example

Digital wallet companies must:

* Protect transaction data
* Avoid unauthorized sharing
* Ensure secure storage

Failure can lead to legal penalties.

---

# **5.5 Avoiding Data Fabrication and Manipulation**

---

## 🔎 Data Fabrication

Making up fake data.

---

## 🔎 Data Manipulation

Altering data to support hypothesis.

---

### ❌ Examples

* Deleting responses that contradict hypothesis
* Changing values to achieve significance

---

### 📌 Academic Consequences

* Research rejection
* Academic penalties
* Loss of credibility

---

# **5.6 Ethical Use of AI and Digital Tools**

With AI tools and automation, ethical responsibility increases.

---

## 📌 Responsible AI Use Includes:

* Transparency in AI-assisted research
* No plagiarism from AI-generated text
* Proper citation
* No automated fake data generation

---

## 🏦 FinTech Risk Example

Using AI to predict credit risk must:

* Avoid bias
* Ensure fairness
* Protect customer data

---

# **5.7 Ethical Issues in Experiments**

When conducting experiments:

* Avoid deception unless justified
* Avoid psychological harm
* Avoid financial loss to participants
* Provide debriefing

---

# **5.8 Ethical Approval (Basic Awareness)**

Some research requires:

* Institutional Review Board (IRB) approval
* Ethics committee clearance

Especially when:

* Collecting sensitive data
* Conducting experiments

---

# 📊 Summary Table

| Ethical Principle | Why Important               |
| ----------------- | --------------------------- |
| Informed Consent  | Protects participant rights |
| Confidentiality   | Protects identity           |
| Privacy           | Protects sensitive data     |
| Data Protection   | Ensures legal compliance    |
| No Fabrication    | Maintains credibility       |
| Responsible AI    | Prevents misuse             |

---

# 🎯 Key Takeaways

* Ethics is non-negotiable in research.
* Financial and personal data require strict protection.
* Consent and confidentiality are essential.
* Fabrication and manipulation destroy credibility.
* AI must be used responsibly.
