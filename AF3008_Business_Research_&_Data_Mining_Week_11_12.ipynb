# 🔹 **1. What is Data Mining?**

## 📌 **1.1 Simple Definition**

👉 **Data Mining means finding useful patterns, trends, or insights from large amounts of data.**

💡 In very simple words:

> **“Data Mining = Finding hidden meaning inside data”**

---

### 🎯 Example:

A bank has data of **1 million customers**.
Data mining helps answer:

* Who is likely to **take a loan**?
* Who might **not repay the loan**?
* Which customers are **high risk**?

---

### 🧠 Key Idea:

* Data is useless if we don’t understand it
* Data mining turns **data → knowledge**

---

## 📌 **1.2 Why Data Mining is Important**

Data today is everywhere but decisions need **insights**, not raw data.

---

### ✅ Reasons why data mining is important:

### 1. Better Decision Making

* Businesses make **smart decisions** using patterns
* Example: Should we give loan or not?

---

### 2. Predict Future Behavior

* Predict what might happen next
* Example:

  * Will customer leave?
  * Will fraud happen?

---

### 3. Save Cost & Increase Profit

* Target the **right customers**
* Avoid losses (e.g., bad loans)

---

### 4. Detect Fraud & Risk

* Banks detect **suspicious transactions**

---

### 🧠 Simple Summary:

> Data Mining helps businesses become **smart, fast, and data-driven**

---

## 📌 **1.3 Real-Life Examples**

---

---

### 🏦 Example 2: Banks

👉 Data Mining Use:

---

### 🛒 Example 3: Amazon

### 📱 Example 4: E-commerce Apps

# 🔹 **2. Data vs Information vs Knowledge**

---

## 📌 **2.1 Raw Data (What it is)**

👉 **Raw Data = Unprocessed facts and numbers**

---

### 🎯 Example:

```
Ali, 25, 50000
Sara, 30, 70000
Ahmed, 22, 20000
```

This is just **data**, no meaning yet.

---

### 🧠 Key Point:

* Data alone is **confusing**
* No decisions can be made directly

---

## 📌 **2.2 Processed Information**

👉 **Information = Organized data that has meaning**

---

### 🎯 Example:

```
Average salary = 46666
Highest salary = 70000
```

Now we understand something.

---

### 🧠 Key Point:

* Data is cleaned and summarized
* Now it becomes useful

---

## 📌 **2.3 Extracted Knowledge (Insights)**

👉 **Knowledge = Actionable understanding from information**

---

### 🎯 Example:

> “Employees with age > 28 have higher salaries”

---

### 🧠 This helps in:

* Decision making
* Strategy building

---

### 🔄 Full Flow:

| Step        | Meaning        |
| ----------- | -------------- |
| Data        | Raw numbers    |
| Information | Organized data |
| Knowledge   | Useful insight |

---

### 💡 Easy Analogy:

* Data = Ingredients 🍅
* Information = Cooked Food 🍲
* Knowledge = Deciding what to eat 🍽️

# 🔹 **3. Data Mining vs Statistics vs Data Analytics**

---

## 📌 **3.1 What is Statistics? (Basic Idea)**

👉 **Statistics is about analyzing data using mathematical formulas**

---

### 🎯 Examples:

* Mean (average)
* Median
* Standard deviation

---

### 🧠 Use:

* Summarize data
* Understand trends

---

### 💡 Simple Example:

Find average marks of students

---

## 📌 **3.2 What is Data Analytics?**

👉 **Data Analytics = Examining data to answer questions and make decisions**

---

### 🎯 Example:

* Why did sales drop last month?
* Which product is best selling?

---

### 🧠 Key Idea:

* Focus on **understanding current and past data**

---

## 📌 **3.3 What is Data Mining?**

👉 **Data Mining = Automatically finding hidden patterns in large data**

---

### 🎯 Example:

* Customers who buy X also buy Y
* Detect fraud patterns

---

### 🧠 Key Idea:

* Uses algorithms
* Works on large datasets
* Focus on discovery

---

## 📌 **3.4 Key Differences (Simple Comparison Table)**

| Feature    | Statistics 📊         | Data Analytics 📈  | Data Mining ⛏️    |
| ---------- | --------------------- | ------------------ | ----------------- |
| Purpose    | Summarize data        | Analyze & explain  | Discover patterns |
| Focus      | Past data             | Past & present     | Hidden insights   |
| Method     | Mathematical formulas | Tools & dashboards | Algorithms & ML   |
| Example    | Average salary        | Sales report       | Fraud detection   |
| Complexity | Simple                | Medium             | Advanced          |

---

### 🧠 Final Understanding:

* **Statistics** → Basic calculations
* **Analytics** → Understanding data
* **Data Mining** → Discovering hidden patterns

---

# ✅ **Quick Recap for Students**

👉 Data Mining = Finding patterns
👉 Data → Information → Knowledge
👉 Statistics vs Analytics vs Mining = Different levels of understanding

# 🔹 **4. Types of Data Mining Tasks**

👉 Data Mining is not just one thing, it performs **different types of tasks depending on the problem**.

---

## 🧠 **Main Idea:**

> Different problems → Different data mining techniques

---

# 📌 **4.1 Classification (Predict Category)**

## ✅ **Simple Definition**

👉 **Classification = Assigning data into predefined categories (labels)**

---

### 🎯 Example:

A bank wants to decide:

👉 Will this customer:

* **Repay Loan ✅**
* **Not Repay Loan ❌**

---

### 📊 Input Data:

* Age
* Income
* Credit history

---

### 📤 Output:

👉 Category (Label):

* Good Customer
* Risky Customer

---

## 🧠 **Key Idea:**

* Output is **fixed categories**
* We already **know possible answers**

---

### 📦 More Real-Life Examples:

* Email → Spam / Not Spam
* Student → Pass / Fail
* Transaction → Fraud / Not Fraud

---

## 🎯 **When to Use Classification?**

✔ When output is **labels or categories**
✔ When decision is **yes/no or type-based**

---

---

# 📌 **4.2 Clustering (Group Similar Items)**

## ✅ **Simple Definition**

👉 **Clustering = Grouping similar data together (without labels)**

---

### 🎯 Example:

An e-commerce company wants to group customers:

👉 Groups might be:

* High spenders 💰
* Medium spenders
* Low spenders

---

### 📊 Input:

* Purchase amount
* Frequency
* Product types

---

### 📤 Output:

👉 Groups (Clusters):

* Cluster 1, Cluster 2, Cluster 3

---

## 🧠 **Key Idea:**

* No predefined labels
* Model finds patterns **on its own**

---

### 📦 Real-Life Examples:

* Customer segmentation
* Market grouping
* Social media user groups

---

## 🎯 **When to Use Clustering?**

✔ When you **don’t know categories beforehand**
✔ When you want to **discover hidden groups**

---

---

# 📌 **4.3 Association (Find Relationships)**

## ✅ **Simple Definition**

👉 **Association = Finding relationships between items**

---

### 🎯 Example:

In a supermarket:

> Customers who buy **bread 🍞** also buy **butter 🧈**

---

### 📊 Input:

* Purchase history

---

### 📤 Output:

👉 Rules like:

* If A → then B

---

## 🧠 **Key Idea:**

* Finds **patterns of co-occurrence**
* Helps in **recommendation systems**

---

### 📦 Real-Life Examples:

* “People who bought this also bought…”
* Product bundling
* Cross-selling

---

## 🎯 **When to Use Association?**

✔ When finding **relationships between items**
✔ When building **recommendation systems**

---

---

# 📌 **4.4 Prediction (Future Forecasting)**

## ✅ **Simple Definition**

👉 **Prediction = Estimating future values based on past data**

---

### 🎯 Example:

A company wants to predict:

👉 Next month sales = ?

---

### 📊 Input:

* Past sales data
* Seasonal trends

---

### 📤 Output:

👉 Numerical value (e.g., 50000 sales)

---

## 🧠 **Key Idea:**

* Output is a **number (continuous value)**
* Focus on **future estimation**

---

### 📦 Real-Life Examples:

* Stock price prediction 📈
* Sales forecasting
* Weather prediction

---

## 🎯 **When to Use Prediction?**

✔ When output is a **number**
✔ When forecasting future trends

---

---

# 🔥 **Final Comparison (Very Important for Students)**

| Task           | What it Does        | Output Type | Example           |
| -------------- | ------------------- | ----------- | ----------------- |
| Classification | Assign category     | Label       | Fraud / Not Fraud |
| Clustering     | Group similar items | Clusters    | Customer groups   |
| Association    | Find relationships  | Rules       | Buy X → Buy Y     |
| Prediction     | Forecast future     | Number      | Sales next month  |

---

---

# 🎯 **Simple Memory Trick**

👉 **C → C → A → P**

* **C**lassification → Categories
* **C**lustering → Groups
* **A**ssociation → Relationships
* **P**rediction → Future

---

---

# ✅ **Quick Recap**

* Classification = Known categories
* Clustering = Unknown groups
* Association = Relationships
* Prediction = Future values

# 🔹 **5. Data Mining Process (Step-by-Step)**

## 🧠 **Main Idea**

> Data Mining is not just applying a model, it is a **complete process from data → decision**

---

## 🎯 **Running Example (We’ll use this in all steps)**

👉 A bank wants to **predict whether a customer will repay a loan or not**

---

# 📌 **5.1 Data Collection**

## ✅ **Definition**

👉 **Collecting raw data from different sources**

---

### 🎯 Example (Bank Case):

Data collected from:

* Customer forms
* Transaction history
* Credit records

---

### 📊 Sample Data:

| Age | Income | Loan | Repayment |
| --- | ------ | ---- | --------- |
| 25  | 50000  | Yes  | Yes       |
| 40  | 30000  | Yes  | No        |

---

## 🧠 Key Point:

* More data = Better results
* Data can come from:

  * Databases
  * APIs
  * Excel files

---

## ⚠️ Problem:

👉 Data may be:

* Incomplete
* Messy
* Inconsistent

---

---

# 📌 **5.2 Data Cleaning**

## ✅ **Definition**

👉 **Fixing errors and handling missing or incorrect data**

---

### 🎯 Example Issues:

| Age | Income | Loan |
| --- | ------ | ---- |
| 25  | NULL   | Yes  |
| -5  | 40000  | Yes  |

---

### 🔧 Cleaning Actions:

* Fill missing values (e.g., average income)
* Remove wrong values (age = -5 ❌)
* Remove duplicates

---

## 🧠 Key Point:

> “Garbage In = Garbage Out” 🚫
> Bad data → Bad results

---

---

# 📌 **5.3 Data Transformation**

## ✅ **Definition**

👉 **Converting data into a suitable format for analysis**

---

### 🎯 Example:

Original Data:

* Income = 50000

Transformed Data:

* Income Category = High

---

### 🔄 Common Transformations:

* Normalization (scaling values)
* Encoding (Yes → 1, No → 0)
* Feature creation (new columns)

---

## 🧠 Key Point:

* Machines understand **numbers better than text**

---

---

# 📌 **5.4 Model Building**

## ✅ **Definition**

👉 **Applying algorithms to learn patterns from data**

---

### 🎯 Example:

Use:

* Decision Tree
* Logistic Regression

---

### 📊 What happens:

* Model learns from past data
* Finds patterns like:

> “If income is low → higher chance of default”

---

## 🧠 Key Point:

* This is where **AI/ML is used**

---

---

# 📌 **5.5 Evaluation**

## ✅ **Definition**

👉 **Checking how well the model performs**

---

### 🎯 Example:

Model predicts:

| Actual | Predicted |
| ------ | --------- |
| Yes    | Yes       |
| No     | Yes ❌     |

---

### 📊 Metrics:

* Accuracy
* Precision
* Recall

---

## 🧠 Key Point:

* We must test before using the model
* Avoid wrong decisions (e.g., giving loan to risky customer)

---

---

# 📌 **5.6 Deployment**

## ✅ **Definition**

👉 **Using the model in real-world systems**

---

### 🎯 Example:

* Bank system automatically:

  * Approves loan
  * Rejects risky customer

---

### 📦 Where models are deployed:

* Mobile apps
* Websites
* Banking systems

---

## 🧠 Key Point:

* This is where business gets value 💰

---

---

# 🔥 **Complete Flow (Very Important)**

👉 Students should remember this:

```
Data → Cleaning → Transformation → Model → Evaluation → Deployment
```

---

# 📌 4.1 What is a Decision Tree?

## ✅ Simple Definition

A **Decision Tree** is a machine learning model that makes decisions by asking a series of **simple questions**.

It starts from one question, then based on the answer, it moves to the next question, and finally reaches a decision.

👉 In simple words:

> A decision tree is like a flowchart that helps us classify something step by step.

---

## ✅ Why is it called a “tree”?

It is called a tree because its structure looks like an upside-down tree:

* It starts from one top point
* Then it branches into smaller paths
* Each path leads to a final result

---

## ✅ Daily Life Analogy

Think about how a human makes decisions.

### Example: Should I take an umbrella?

You may think like this:

* Is it cloudy?

  * If no → No umbrella
  * If yes → Is rain expected?

    * If yes → Take umbrella
    * If no → No umbrella

This is exactly how a decision tree works.

It asks one question at a time until it reaches an answer.

---

## ✅ In Machine Learning

In classification, a decision tree uses input features such as:

* age
* income
* marks
* attendance
* credit score
* purchase frequency

and then decides the class or category, such as:

* Pass / Fail
* Fraud / Not Fraud
* Approve / Reject
* High Risk / Low Risk

---

## ✅ Key Idea

A decision tree learns rules from old data.

For example, it may learn rules like:

* If income is high and credit history is good → Approve loan
* If income is low and credit history is poor → Reject loan

So instead of a human writing all rules manually, the model learns these rules from data.

---

# 📌 4.2 Tree Structure (Root, Nodes, Leaves)

To understand decision trees properly, students must understand the three main parts:

1. Root Node
2. Internal Nodes
3. Leaf Nodes

---

## ✅ 1. Root Node

The **root node** is the starting point of the tree.

It is the first and most important question.

### Example:

* Is income greater than 50,000?

This is where the tree begins.

---

## ✅ 2. Internal Nodes

The **internal nodes** are the middle questions.

These nodes split the data further based on conditions.

### Example:

* Is credit score good?
* Is age above 30?
* Is previous loan default = No?

These questions help narrow down the decision.

---

## ✅ 3. Leaf Nodes

The **leaf nodes** are the final outputs or decisions.

They do not ask more questions.

They give the final class.

### Example:

* Loan Approved
* Loan Rejected
* Fraud
* Not Fraud

---

## ✅ Simple Visual Understanding

```text
                 [Income > 50000?]
                    /         \
                 Yes           No
                /               \
      [Credit Score Good?]     Reject
          /        \
        Yes        No
       /            \
   Approve         Reject
```

---

## ✅ Meaning of this tree

This tree says:

* First check income
* If income is low, reject directly
* If income is high, then check credit score
* If credit score is good, approve
* Otherwise reject

This is exactly how decision trees make decisions.

---

## ✅ Important Terms

| Term   | Meaning                           |
| ------ | --------------------------------- |
| Root   | Starting question                 |
| Node   | Decision/question point           |
| Branch | Path from one question to another |
| Leaf   | Final decision/output             |

---

# 📌 4.5 Limitations

Although decision trees are very useful, they are not perfect.

Students should also understand their weaknesses.

---

## ✅ 1. Can Overfit the Data

A decision tree may become too detailed and memorize training data instead of learning general patterns.

This is called **overfitting**.

### Meaning:

The model performs very well on old data but poorly on new unseen data.

### Example:

If the tree keeps making too many tiny branches, it may become too specific.

---

## ✅ 2. Small Changes in Data Can Change the Tree

If training data changes slightly, the structure of the tree may change a lot.

This means decision trees can sometimes be unstable.

---

## ✅ 3. Can Become Very Large and Complex

For big datasets, the tree may become too deep and difficult to read.

A very large tree loses the simplicity that makes decision trees attractive.

---

## ✅ 4. May Not Always Give Highest Accuracy

Decision trees are easy to understand, but sometimes other models such as random forests or boosting models perform better in accuracy.

So decision trees are great for explanation, but not always the best performer.

---

## ✅ 5. Biased if Data is Imbalanced

If one class appears much more than another, the tree may favor the majority class.

### Example:

If 95% transactions are normal and only 5% are fraud, the tree may mostly predict “not fraud”.

---

## ✅ 6. Greedy Nature

Decision trees often choose the best split at the current step, but that choice may not be the best overall long-term choice.

This is why decision trees are sometimes called greedy algorithms.

---

## ✅ Summary of Limitations

| Limitation              | Explanation                               |
| ----------------------- | ----------------------------------------- |
| Overfitting             | Learns too much detail from training data |
| Instability             | Small data change can change the tree     |
| Complexity              | Large trees become hard to read           |
| Accuracy issue          | Other models may perform better           |
| Class imbalance problem | Can favor majority class                  |

# 🔹 1. What is a Decision Tree?

# 🔹 5. Important Definitions

## 5.1 Entropy

Entropy tells us how mixed or impure the dataset is.

### ✅ Simple meaning

* If all records belong to one class only, entropy is **0**
* If records are mixed, entropy becomes larger
* More mixing = more uncertainty

### ✅ Formula

$$
Entropy(S) = -p_1 \log_2(p_1) - p_2 \log_2(p_2)
$$

Where:

* $$p_1$$   = probability of class 1
* $$p_2$$   = probability of class 2

In our example:

* class 1 = Yes
* class 2 = No

---

## 5.2 Information Gain

Information Gain tells us how much better a split makes the data.

### ✅ Simple meaning

It measures how much uncertainty is reduced after splitting the data.

### ✅ Formula

$$
IG(S, A) = Entropy(S) - Entropy_{after}(A)
$$

Where:

* $$S$$ = full dataset
* $$A$$ = feature used for splitting

---

## 5.3 Weighted Entropy After Split

When we split data into groups, we calculate the entropy of each group and combine them.

### ✅ Formula

$$
Entropy_{after}(A) =
\sum_{i=1}^{n}
\frac{|S_i|}{|S|} \times Entropy(S_i)
$$

Where:

* $$|S_i|$$ = number of records in subset $$i$$
* $$|S|$$ = total number of records

---

# 📘 Decision Tree Complete Example

This notebook demonstrates a **Decision Tree** step by step for a **Loan Approval** problem.

## 🎯 Goal
Predict whether a loan should be:

- **Yes** = Approve
- **No** = Reject

## 📌 Features Used
- **Income**
- **Credit Score**

## 📌 What this code shows
- Original dataset
- Entropy of the full dataset
- Information Gain for each feature
- Best root node selection
- Second split of the tree
- Final tree visualization
- Rules extracted from the tree
- Predictions for new customers

````python
# ============================================================
# DECISION TREE - COMPLETE BEGINNER FRIENDLY STEP-BY-STEP DEMO
# Loan Approval Example
# ============================================================

import math
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# -------------------------------
# 1. CREATE THE DATASET
# -------------------------------
data = {
    "Customer": ["C1", "C2", "C3", "C4", "C5", "C6", "C7", "C8"],
    "Income": ["High", "High", "High", "Medium", "Medium", "Low", "Low", "Low"],
    "Credit Score": ["Good", "Good", "Poor", "Good", "Poor", "Good", "Poor", "Poor"],
    "Loan Approved": ["Yes", "Yes", "No", "Yes", "No", "No", "No", "No"]
}

df = pd.DataFrame(data)

print("=" * 70)
print("STEP 1: ORIGINAL DATASET")
print("=" * 70)
print(df.to_string(index=False))


# -------------------------------
# 2. HELPER FUNCTIONS
# -------------------------------
def entropy_from_list(labels):
    """
    Calculate entropy from a list of class labels.
    """
    total = len(labels)
    counts = Counter(labels)

    ent = 0
    for label, count in counts.items():
        p = count / total
        if p > 0:
            ent -= p * math.log2(p)
    return ent


def show_entropy_details(labels, title="Entropy Calculation"):
    """
    Show step-by-step entropy calculation.
    """
    total = len(labels)
    counts = Counter(labels)

    print("\n" + "-" * 70)
    print(title)
    print("-" * 70)
    print(f"Total records = {total}")

    ent = 0
    for label, count in counts.items():
        p = count / total
        print(f"Class '{label}' count = {count}")
        print(f"p({label}) = {count}/{total} = {p:.4f}")
        if p > 0:
            term = -p * math.log2(p)
            ent += term
            print(f"Contribution of class '{label}' = -({p:.4f}) * log2({p:.4f}) = {term:.4f}")

    print(f"\nFinal Entropy = {ent:.4f}")
    return ent


def information_gain(df, feature, target):
    """
    Calculate information gain for a given feature.
    Also print step-by-step details.
    """
    print("\n" + "=" * 70)
    print(f"CALCULATING INFORMATION GAIN FOR FEATURE: {feature}")
    print("=" * 70)

    # Entropy before split
    parent_entropy = entropy_from_list(df[target])
    print(f"Entropy of full dataset = {parent_entropy:.4f}")

    total_records = len(df)
    weighted_entropy = 0

    print(f"\nSplitting dataset using feature: {feature}")
    unique_values = df[feature].unique()

    for value in unique_values:
        subset = df[df[feature] == value]
        subset_entropy = entropy_from_list(subset[target])
        weight = len(subset) / total_records
        weighted_part = weight * subset_entropy

        print("\n" + "-" * 50)
        print(f"Subset where {feature} = {value}")
        print("-" * 50)
        print(subset.to_string(index=False))
        print(f"Subset size = {len(subset)}")
        print(f"Weight = {len(subset)}/{total_records} = {weight:.4f}")
        print(f"Subset entropy = {subset_entropy:.4f}")
        print(f"Weighted entropy contribution = {weight:.4f} * {subset_entropy:.4f} = {weighted_part:.4f}")

        weighted_entropy += weighted_part

    ig = parent_entropy - weighted_entropy

    print("\n" + "-" * 50)
    print(f"Weighted Entropy after splitting on '{feature}' = {weighted_entropy:.4f}")
    print(f"Information Gain = {parent_entropy:.4f} - {weighted_entropy:.4f} = {ig:.4f}")

    return ig, weighted_entropy


def majority_class(labels):
    """
    Return the majority class in a list of labels.
    """
    return Counter(labels).most_common(1)[0][0]


# -------------------------------
# 3. CALCULATE ENTROPY OF FULL DATASET
# -------------------------------
full_entropy = show_entropy_details(df["Loan Approved"], title="STEP 2: ENTROPY OF FULL DATASET")


# -------------------------------
# 4. CALCULATE INFORMATION GAIN FOR EACH FEATURE
# -------------------------------
ig_income, weighted_income = information_gain(df, "Income", "Loan Approved")
ig_credit, weighted_credit = information_gain(df, "Credit Score", "Loan Approved")

print("\n" + "=" * 70)
print("STEP 3: COMPARISON OF FEATURES")
print("=" * 70)
print(f"Information Gain of Income       = {ig_income:.4f}")
print(f"Information Gain of Credit Score = {ig_credit:.4f}")

if ig_credit > ig_income:
    root_feature = "Credit Score"
else:
    root_feature = "Income"

print(f"\nBest feature for ROOT node = {root_feature}")


# -------------------------------
# 5. VISUALIZE INFORMATION GAIN
# -------------------------------
features = ["Income", "Credit Score"]
gains = [ig_income, ig_credit]

plt.figure(figsize=(8, 5))
bars = plt.bar(features, gains)
plt.title("Information Gain Comparison")
plt.xlabel("Feature")
plt.ylabel("Information Gain")

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.01, f"{height:.3f}",
             ha='center', va='bottom', fontsize=10)

plt.ylim(0, max(gains) + 0.15)
plt.show()


# -------------------------------
# 6. BUILD THE TREE MANUALLY STEP BY STEP
#    (BEGINNER FRIENDLY FOR THIS SMALL EXAMPLE)
# -------------------------------
print("\n" + "=" * 70)
print("STEP 4: BUILDING THE DECISION TREE")
print("=" * 70)

print(f"\nROOT NODE = {root_feature}")

# Since Credit Score is the best root in this example:
# Split the data by Credit Score
root_groups = {}
for value in df[root_feature].unique():
    root_groups[value] = df[df[root_feature] == value]

for group_name, group_df in root_groups.items():
    print("\n" + "-" * 70)
    print(f"BRANCH: {root_feature} = {group_name}")
    print("-" * 70)
    print(group_df.to_string(index=False))

    group_entropy = entropy_from_list(group_df["Loan Approved"])
    print(f"Entropy of this branch = {group_entropy:.4f}")

    # If pure node
    unique_classes = group_df["Loan Approved"].unique()
    if len(unique_classes) == 1:
        print(f"This branch is PURE.")
        print(f"Final decision = {unique_classes[0]}")
    else:
        print("This branch is NOT pure.")
        print("We need another split.")

        # Remaining feature for the second split
        remaining_feature = "Income" if root_feature == "Credit Score" else "Credit Score"
        second_ig, _ = information_gain(group_df, remaining_feature, "Loan Approved")
        print(f"\nBest next split for this branch = {remaining_feature}")

        # Split again
        second_groups = group_df.groupby(remaining_feature)
        for second_value, second_df in second_groups:
            print("\n" + " " * 5 + f"Sub-branch: {remaining_feature} = {second_value}")
            print(" " * 5 + "-" * 45)
            print(second_df.to_string(index=False))

            classes_here = second_df["Loan Approved"].unique()
            if len(classes_here) == 1:
                print(" " * 5 + f"Final decision = {classes_here[0]}")
            else:
                maj = majority_class(second_df["Loan Approved"])
                print(" " * 5 + f"Still mixed, majority class = {maj}")


# -------------------------------
# 7. FINAL TREE VISUALIZATION (TEXT)
# -------------------------------
print("\n" + "=" * 70)
print("STEP 5: FINAL DECISION TREE")
print("=" * 70)

tree_text = r"""
                    [Credit Score?]
                     /           \
                 Good             Poor
                  /                 \
             [Income?]             Reject
             /   |    \
         High  Medium  Low
          |      |      |
       Approve Approve Reject
"""
print(tree_text)


# -------------------------------
# 8. VISUALIZE THE TREE USING MATPLOTLIB TEXT
# -------------------------------
plt.figure(figsize=(10, 6))
plt.axis("off")

# Root
plt.text(0.5, 0.9, "Credit Score?", ha="center", va="center",
         bbox=dict(boxstyle="round,pad=0.4", fc="lightgray", ec="black"))

# Level 1
plt.text(0.3, 0.65, "Income?", ha="center", va="center",
         bbox=dict(boxstyle="round,pad=0.4", fc="lightblue", ec="black"))
plt.text(0.7, 0.65, "Reject", ha="center", va="center",
         bbox=dict(boxstyle="round,pad=0.4", fc="lightcoral", ec="black"))

# Level 2
plt.text(0.15, 0.35, "Approve", ha="center", va="center",
         bbox=dict(boxstyle="round,pad=0.4", fc="lightgreen", ec="black"))
plt.text(0.30, 0.35, "Approve", ha="center", va="center",
         bbox=dict(boxstyle="round,pad=0.4", fc="lightgreen", ec="black"))
plt.text(0.45, 0.35, "Reject", ha="center", va="center",
         bbox=dict(boxstyle="round,pad=0.4", fc="lightcoral", ec="black"))

# Lines
plt.plot([0.5, 0.3], [0.87, 0.69], 'k-')
plt.plot([0.5, 0.7], [0.87, 0.69], 'k-')

plt.plot([0.3, 0.15], [0.62, 0.39], 'k-')
plt.plot([0.3, 0.30], [0.62, 0.39], 'k-')
plt.plot([0.3, 0.45], [0.62, 0.39], 'k-')

# Branch labels
plt.text(0.38, 0.80, "Good", fontsize=10)
plt.text(0.60, 0.80, "Poor", fontsize=10)

plt.text(0.20, 0.50, "High", fontsize=10)
plt.text(0.30, 0.50, "Medium", fontsize=10)
plt.text(0.39, 0.50, "Low", fontsize=10)

plt.title("Final Decision Tree Visualization", pad=20)
plt.show()


# -------------------------------
# 9. EXTRACT RULES
# -------------------------------
print("\n" + "=" * 70)
print("STEP 6: RULES EXTRACTED FROM THE TREE")
print("=" * 70)

print("Rule 1: If Credit Score = Poor, then Loan Approved = No")
print("Rule 2: If Credit Score = Good and Income = High, then Loan Approved = Yes")
print("Rule 3: If Credit Score = Good and Income = Medium, then Loan Approved = Yes")
print("Rule 4: If Credit Score = Good and Income = Low, then Loan Approved = No")


# -------------------------------
# 10. MANUAL PREDICTION FUNCTION
# -------------------------------
def predict_loan(income, credit_score):
    """
    Predict using the final manually built decision tree.
    """
    if credit_score == "Poor":
        return "No"
    else:  # Good
        if income == "High":
            return "Yes"
        elif income == "Medium":
            return "Yes"
        elif income == "Low":
            return "No"
        else:
            return "Unknown"


# -------------------------------
# 11. TEST ON NEW CUSTOMERS
# -------------------------------
print("\n" + "=" * 70)
print("STEP 7: TEST THE TREE ON NEW EXAMPLES")
print("=" * 70)

test_cases = [
    {"Income": "High", "Credit Score": "Good"},
    {"Income": "Medium", "Credit Score": "Poor"},
    {"Income": "Low", "Credit Score": "Good"},
    {"Income": "High", "Credit Score": "Poor"},
]

results = []
for case in test_cases:
    pred = predict_loan(case["Income"], case["Credit Score"])
    results.append({
        "Income": case["Income"],
        "Credit Score": case["Credit Score"],
        "Predicted Loan Approval": pred
    })

test_df = pd.DataFrame(results)
print(test_df.to_string(index=False))


# -------------------------------
# 12. SHOW CLASS DISTRIBUTION
# -------------------------------
print("\n" + "=" * 70)
print("STEP 8: CLASS DISTRIBUTION IN ORIGINAL DATASET")
print("=" * 70)

class_counts = df["Loan Approved"].value_counts()
print(class_counts)

plt.figure(figsize=(6, 4))
bars = plt.bar(class_counts.index, class_counts.values)
plt.title("Loan Approval Class Distribution")
plt.xlabel("Loan Approved")
plt.ylabel("Count")

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.05, str(height),
             ha='center', va='bottom')

plt.ylim(0, max(class_counts.values) + 1)
plt.show()


# -------------------------------
# 13. FINAL SUMMARY
# -------------------------------
print("\n" + "=" * 70)
print("STEP 9: FINAL SUMMARY")
print("=" * 70)
print(f"Initial entropy of dataset        = {full_entropy:.4f}")
print(f"Information Gain of Income        = {ig_income:.4f}")
print(f"Information Gain of Credit Score  = {ig_credit:.4f}")
print(f"Best root node                    = {root_feature}")

print("\nConclusion:")
print("The tree first splits on Credit Score because it has the highest Information Gain.")
print("Then, for customers with Good Credit Score, it splits on Income.")
print("This creates the final decision rules for loan approval.")
````

# 📘 Decision Tree using Scikit-Learn

In this example, we build a **Decision Tree Classifier** for a **Loan Approval** problem.

## 🎯 Goal
Predict whether a loan should be:

- **Yes** = Approve
- **No** = Reject

## 📌 Features Used
- Income
- Credit Score

## 📌 What students will learn
- how to prepare categorical data for machine learning
- how to train a decision tree using `DecisionTreeClassifier`
- how to visualize the decision tree
- how to interpret feature importance
- how to predict new examples

````python
# ============================================================
# DECISION TREE USING SCIKIT-LEARN (BEGINNER FRIENDLY)
# Loan Approval Example
# ============================================================

# -------------------------------
# 1. IMPORT LIBRARIES
# -------------------------------
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# -------------------------------
# 2. CREATE THE DATASET
# -------------------------------
data = {
    "Customer": ["C1", "C2", "C3", "C4", "C5", "C6", "C7", "C8"],
    "Income": ["High", "High", "High", "Medium", "Medium", "Low", "Low", "Low"],
    "Credit Score": ["Good", "Good", "Poor", "Good", "Poor", "Good", "Poor", "Poor"],
    "Loan Approved": ["Yes", "Yes", "No", "Yes", "No", "No", "No", "No"]
}

df = pd.DataFrame(data)

print("=" * 70)
print("STEP 1: ORIGINAL DATASET")
print("=" * 70)
print(df.to_string(index=False))

# -------------------------------
# 3. EXPLAIN THE PROBLEM
# -------------------------------
print("\n" + "=" * 70)
print("STEP 2: PROBLEM UNDERSTANDING")
print("=" * 70)
print("We want to train a Decision Tree model.")
print("Input features  : Income, Credit Score")
print("Target output   : Loan Approved (Yes/No)")

# -------------------------------
# 4. ENCODE CATEGORICAL DATA
# -------------------------------
# Machine learning models work with numbers, not text.
# So we convert text values into numeric form.

income_encoder = LabelEncoder()
credit_encoder = LabelEncoder()
target_encoder = LabelEncoder()

df["Income_Encoded"] = income_encoder.fit_transform(df["Income"])
df["Credit_Encoded"] = credit_encoder.fit_transform(df["Credit Score"])
df["Loan_Encoded"] = target_encoder.fit_transform(df["Loan Approved"])

print("\n" + "=" * 70)
print("STEP 3: ENCODED DATASET")
print("=" * 70)
print(df[["Customer", "Income", "Income_Encoded", "Credit Score", "Credit_Encoded", "Loan Approved", "Loan_Encoded"]].to_string(index=False))

print("\nEncoding Meaning:")
print("Income Encoding:")
for text, num in zip(income_encoder.classes_, range(len(income_encoder.classes_))):
    print(f"  {text} -> {num}")

print("\nCredit Score Encoding:")
for text, num in zip(credit_encoder.classes_, range(len(credit_encoder.classes_))):
    print(f"  {text} -> {num}")

print("\nLoan Approved Encoding:")
for text, num in zip(target_encoder.classes_, range(len(target_encoder.classes_))):
    print(f"  {text} -> {num}")

# -------------------------------
# 5. PREPARE FEATURES AND TARGET
# -------------------------------
X = df[["Income_Encoded", "Credit_Encoded"]]
y = df["Loan_Encoded"]

print("\n" + "=" * 70)
print("STEP 4: FEATURES (X) AND TARGET (y)")
print("=" * 70)
print("Features used for training:")
print(X.to_string(index=False))

print("\nTarget values:")
print(y.to_string(index=False))

# -------------------------------
# 6. CREATE AND TRAIN THE MODEL
# -------------------------------
# criterion='entropy' tells sklearn to use entropy / information gain
# random_state is used to make results reproducible

model = DecisionTreeClassifier(criterion="entropy", random_state=42)
model.fit(X, y)

print("\n" + "=" * 70)
print("STEP 5: MODEL TRAINED SUCCESSFULLY")
print("=" * 70)
print("A Decision Tree has been trained using scikit-learn.")

# -------------------------------
# 7. PREDICT ON TRAINING DATA
# -------------------------------
y_pred = model.predict(X)

print("\n" + "=" * 70)
print("STEP 6: PREDICTIONS ON TRAINING DATA")
print("=" * 70)

results_df = df[["Customer", "Income", "Credit Score", "Loan Approved"]].copy()
results_df["Predicted"] = target_encoder.inverse_transform(y_pred)
print(results_df.to_string(index=False))

# -------------------------------
# 8. EVALUATE THE MODEL
# -------------------------------
accuracy = accuracy_score(y, y_pred)

print("\n" + "=" * 70)
print("STEP 7: MODEL EVALUATION")
print("=" * 70)
print(f"Accuracy = {accuracy:.2f}")

print("\nConfusion Matrix:")
cm = confusion_matrix(y, y_pred)
print(cm)

print("\nClassification Report:")
print(classification_report(y, y_pred, target_names=target_encoder.classes_))

# -------------------------------
# 9. VISUALIZE THE DECISION TREE
# -------------------------------
plt.figure(figsize=(14, 8))
plot_tree(
    model,
    feature_names=["Income", "Credit Score"],
    class_names=target_encoder.classes_,
    filled=True,
    rounded=True,
    fontsize=11
)
plt.title("Decision Tree Visualization")
plt.show()

# -------------------------------
# 10. SHOW FEATURE IMPORTANCE
# -------------------------------
print("\n" + "=" * 70)
print("STEP 8: FEATURE IMPORTANCE")
print("=" * 70)

importance_df = pd.DataFrame({
    "Feature": ["Income", "Credit Score"],
    "Importance": model.feature_importances_
})

print(importance_df.to_string(index=False))

plt.figure(figsize=(6, 4))
bars = plt.bar(importance_df["Feature"], importance_df["Importance"])
plt.title("Feature Importance")
plt.xlabel("Features")
plt.ylabel("Importance Score")

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.01, f"{height:.3f}",
             ha='center', va='bottom')

plt.ylim(0, max(importance_df["Importance"]) + 0.2)
plt.show()

# -------------------------------
# 11. INTERPRET THE TREE IN TEXT FORM
# -------------------------------
print("\n" + "=" * 70)
print("STEP 9: SIMPLE INTERPRETATION")
print("=" * 70)
print("The Decision Tree automatically learned rules from the data.")
print("It splits the data using the most informative feature first.")
print("The tree diagram above shows:")
print("- the first question asked by the model")
print("- the next questions asked in each branch")
print("- the final prediction at the leaf nodes")

# -------------------------------
# 12. TEST THE MODEL ON NEW CUSTOMERS
# -------------------------------
new_customers = pd.DataFrame({
    "Income": ["High", "Medium", "Low", "High"],
    "Credit Score": ["Good", "Poor", "Good", "Poor"]
})

print("\n" + "=" * 70)
print("STEP 10: NEW CUSTOMERS")
print("=" * 70)
print(new_customers.to_string(index=False))

# Encode new data using the same encoders
new_customers["Income_Encoded"] = income_encoder.transform(new_customers["Income"])
new_customers["Credit_Encoded"] = credit_encoder.transform(new_customers["Credit Score"])

X_new = new_customers[["Income_Encoded", "Credit_Encoded"]]
new_predictions = model.predict(X_new)

new_customers["Predicted Loan Approval"] = target_encoder.inverse_transform(new_predictions)

print("\nPredictions for New Customers:")
print(new_customers[["Income", "Credit Score", "Predicted Loan Approval"]].to_string(index=False))

# -------------------------------
# 13. VISUALIZE PREDICTION COUNTS
# -------------------------------
pred_counts = new_customers["Predicted Loan Approval"].value_counts()

plt.figure(figsize=(6, 4))
bars = plt.bar(pred_counts.index, pred_counts.values)
plt.title("Prediction Counts for New Customers")
plt.xlabel("Predicted Class")
plt.ylabel("Count")

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.05, str(height),
             ha='center', va='bottom')

plt.ylim(0, max(pred_counts.values) + 1)
plt.show()

# -------------------------------
# 14. FINAL SUMMARY
# -------------------------------
print("\n" + "=" * 70)
print("STEP 11: FINAL SUMMARY")
print("=" * 70)
print("1. We created a small loan approval dataset.")
print("2. We converted text values into numbers using LabelEncoder.")
print("3. We trained a DecisionTreeClassifier using entropy.")
print("4. We visualized the tree.")
print("5. We checked feature importance.")
print("6. We used the trained model to predict new customers.")

print("\nThis is how Decision Trees are built using scikit-learn.")
````
