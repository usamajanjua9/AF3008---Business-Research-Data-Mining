# **Week 7 - Sampling & Measurement**

# **1. Introduction to Sampling**

Sampling is a crucial concept in business research because studying an entire population is often impossible or impractical.

---

# **1.1 Meaning of Population in Research**

---

## 🔎 What is a Population?

In research, a **population** refers to the complete group of individuals, objects, or events that the researcher wants to study.

---

### 🧠 Simple Explanation

> Population = The entire group you want to understand.

---

### 📌 Examples in Business Research

* All university students in Islamabad
* All customers of PaySecure
* All bank account holders in Pakistan
* All SMEs in Lahore

---

### 🏦 FinTech Example

If PaySecure wants to study digital wallet adoption among students:

Population = All university students in Pakistan.

---

# **1.2 Definition of Target Population**

---

## 🔎 What is Target Population?

Target population is the **specific group within the broader population** that the researcher actually intends to study.

---

### 🧠 Difference from Population

Population = All possible units
Target Population = Specific group relevant to study

---

### 🏦 Example

Population: All smartphone users in Pakistan
Target Population: University students using digital wallets in Islamabad

---

Target population must match research objectives.

---

# **1.3 Sampling Frame**

---

## 🔎 What is a Sampling Frame?

A sampling frame is the **actual list or database from which a sample is selected**.

It is the operational representation of the target population.

---

### 🧠 Simple Explanation

> Sampling frame = The list you use to select participants.

---

### 📌 Examples

* University student enrollment list
* Customer database
* Voter registration list
* CRM database

---

### 🏦 FinTech Example

If studying wallet usage:

Sampling frame = Registered PaySecure users database.

---

### ⚠ Important

If sampling frame is incomplete → results may be biased.

Example:
If only active users are listed → non-users are excluded.

---

# **1.4 Need for Sampling in Business Research**

---

## 🔎 Why Not Study the Entire Population?

Studying entire population (census) may be:

* Expensive
* Time-consuming
* Logistically impossible
* Operationally inefficient

---

### 🧠 Simple Explanation

> Sampling saves time, money, and effort.

---

### 🏦 Example

PaySecure has 2 million users.

Surveying all 2 million users is unrealistic.

Instead, survey 500–1000 representative users.

---

### 📌 Benefits of Sampling

* Cost efficiency
* Faster results
* Easier data analysis
* Practical feasibility

---

# **1.5 Census vs Sampling**

---

## 🔎 Census

A census collects data from **every member of the population**.

---

### 📌 Example

National population census by government.

---

## 🔎 Sampling

Sampling collects data from **a subset of the population**.

---

### 📊 Comparison Table

| Feature     | Census            | Sampling                  |
| ----------- | ----------------- | ------------------------- |
| Coverage    | Entire population | Subset only               |
| Cost        | High              | Low                       |
| Time        | Long              | Short                     |
| Feasibility | Often difficult   | Practical                 |
| Accuracy    | High (if perfect) | Depends on sample quality |

---

### 🏦 Business Example

Census:
Analyze all 2 million transactions.

Sampling:
Analyze 10,000 randomly selected transactions.

---

# **1.6 Sampling Error**

---

## 🔎 What is Sampling Error?

Sampling error is the difference between:

* Sample results
  AND
* Actual population value

---

### 🧠 Simple Explanation

> Because we study only a part of the population, results may not be perfectly accurate.

---

### 📌 Example

True population wallet usage = 70%
Sample result = 65%

Difference (5%) = Sampling error

---

## 📌 Why Sampling Error Occurs

* Random variation
* Small sample size
* Poor sampling technique

---

## 📌 How to Reduce Sampling Error

* Increase sample size
* Use probability sampling
* Ensure representative sampling frame

---

# 📊 Visual Understanding

Population → Sampling Frame → Sample → Analysis → Conclusion

---

# 🎯 Key Takeaways

* Population = entire group of interest.
* Target population = specific research focus group.
* Sampling frame = list used to select participants.
* Sampling is necessary due to practical constraints.
* Sampling error is unavoidable but manageable.

# **2. Population and Sampling Concepts**

Understanding these concepts ensures that research findings are scientifically sound and statistically valid.

---

# **2.1 Population Parameters vs Sample Statistics**

---

## 🔎 Population Parameter

A **population parameter** is a numerical value that describes a characteristic of the entire population.

Examples:

* Population mean income
* Population percentage of wallet users
* Population average transaction value

Parameters are usually represented using Greek symbols:

* μ (population mean)
* σ (population standard deviation)
* p (population proportion)

---

## 🔎 Sample Statistic

A **sample statistic** is a numerical value calculated from a sample.

Examples:

* Sample mean income
* Sample percentage using digital wallets
* Sample average transaction value

Statistics are represented using:

* x̄ (sample mean)
* s (sample standard deviation)
* p̂ (sample proportion)

---

### 🧠 Simple Explanation

> Parameter = True value (unknown)
> Statistic = Estimated value (calculated from sample)

---

### 🏦 Example

True wallet usage in population = 70% (parameter)
Survey result shows 65% (statistic)

Statistic estimates the parameter.

---

# **2.2 Representativeness**

---

## 🔎 What is Representativeness?

A sample is representative if it accurately reflects the characteristics of the population.

---

### 📌 Why It Matters

If sample is not representative → results will be misleading.

---

### 🏦 Example

Population:
50% male, 50% female students

Sample:
90% male, 10% female

This sample is not representative.

---

## 📌 How to Ensure Representativeness

* Use probability sampling
* Increase sample size
* Ensure diversity
* Use stratified sampling

---

# **2.3 Sampling Bias**

---

## 🔎 What is Sampling Bias?

Sampling bias occurs when some members of the population have a higher or lower chance of being selected.

---

### 🧠 Simple Explanation

> Sampling bias happens when sample does not fairly represent population.

---

### 🏦 Example

Survey only students in business department → excludes engineering students.

This creates bias.

---

### 📌 Common Types of Bias

* Selection bias
* Non-response bias
* Convenience bias

---

# **2.4 Sample Size Considerations**

---

## 🔎 Why Sample Size Matters

Larger sample size:

* Reduces sampling error
* Improves reliability
* Improves generalizability

---

### 📌 Small Sample Problems

* High variability
* Less accurate estimates
* Low statistical power

---

### 🏦 Example

Surveying 20 students vs 400 students:

400 students → more reliable results.

---

## 📌 Factors Affecting Sample Size

* Population size
* Desired confidence level
* Margin of error
* Research design
* Budget and time constraints

---

# **2.5 Generalizability**

---

## 🔎 What is Generalizability?

Generalizability means:

> The extent to which research findings apply to the larger population.

---

### 📌 Example

If survey is conducted only in Islamabad:

Can results apply to entire Pakistan?

Depends on representativeness.

---

### 📌 Factors Affecting Generalizability

* Sampling technique
* Sample size
* Population diversity
* Research design

---

# 🎯 Key Takeaways (Section 2)

* Population parameter is true value; statistic is estimate.
* Representativeness is critical.
* Sampling bias distorts results.
* Larger samples improve accuracy.
* Good sampling improves generalizability.

---

# **3. Probability Sampling Techniques**

Probability sampling ensures every member of the population has a known and non-zero chance of selection.

This reduces bias and improves generalizability.

---

# **3.1 Simple Random Sampling**

---

## 🔎 Definition

Every member of population has equal probability of being selected.

---

### 📌 Method

* Assign numbers to population members
* Use random number generator

---

### 🏦 Example

Select 200 students randomly from university enrollment list.

---

## ✅ Advantages

* Simple
* Reduces bias
* Easy statistical analysis

---

## ❌ Limitations

* Requires complete sampling frame
* May not represent subgroups proportionally

---

# **3.2 Systematic Sampling**

---

## 🔎 Definition

Select every k-th element from population list.

k = Population size / Sample size

---

### 🏦 Example

Population = 2000 students
Sample = 200

k = 2000 / 200 = 10

Select every 10th student.

---

## ✅ Advantages

* Easy to implement
* Efficient

---

## ❌ Limitations

* Risk of hidden pattern bias
* Not fully random if pattern exists

---

# **3.3 Stratified Sampling**

---

## 🔎 Definition

Divide population into subgroups (strata) and sample from each group.

---

### 🧠 Purpose

Ensure representation of important subgroups.

---

### 🏦 Example

Stratify by:

* Gender
* Department
* Income level

If population:

60% male
40% female

Sample should maintain same ratio.

---

## ✅ Advantages

* Highly representative
* Reduces sampling error
* Useful in diverse populations

---

## ❌ Limitations

* Requires knowledge of population characteristics
* More complex

---

# **3.4 Cluster Sampling**

---

## 🔎 Definition

Divide population into clusters and randomly select entire clusters.

---

### 🏦 Example

Population = All universities in Pakistan

Randomly select 5 universities
Survey all students in selected universities.

---

## ✅ Advantages

* Cost-effective
* Useful for geographically dispersed populations

---

## ❌ Limitations

* Higher sampling error
* Less precise than stratified sampling

---

# **3.5 Advantages and Limitations of Probability Sampling**

---

## ✅ Advantages

* Reduces bias
* Allows statistical inference
* Supports hypothesis testing
* Improves generalizability

---

## ❌ Limitations

* Requires complete sampling frame
* Can be expensive
* May be difficult in large populations

---

# 📊 Summary Table

| Method        | Best For          | Strength              | Weakness        |
| ------------- | ----------------- | --------------------- | --------------- |
| Simple Random | Small populations | Fair & unbiased       | Needs full list |
| Systematic    | Organized lists   | Easy                  | Pattern risk    |
| Stratified    | Diverse groups    | Highly representative | Complex         |
| Cluster       | Large populations | Cost-effective        | Higher error    |

---

# 🎯 Final Takeaways

* Probability sampling supports scientific research.
* Stratified sampling is often best for business research.
* Sample quality determines research credibility.
* Proper sampling improves generalizability.

---





# **4. Non-Probability Sampling Techniques**

Unlike probability sampling, non-probability sampling does **not give every population member an equal chance of selection**.

It is commonly used in:

* Exploratory research
* Time-limited projects
* Hard-to-reach populations
* Pilot studies

---

## **4.1 Convenience Sampling**

---

### 🔎 Definition

Convenience sampling selects participants who are **easily accessible**.

---

### 🧠 Simple Explanation

> Choose whoever is easiest to reach.

---

### 🏦 Example

Survey students in your own classroom.

---

### ✅ Advantages

* Quick
* Low cost
* Easy to implement

---

### ❌ Limitations

* High bias risk
* Low representativeness
* Poor generalizability

---

### 📌 When Appropriate?

* Preliminary research
* Pilot testing questionnaire

---

## **4.2 Judgmental (Purposive) Sampling**

---

### 🔎 Definition

Researcher selects participants based on their **expertise or relevance**.

---

### 🧠 Simple Explanation

> Select people who are most useful for your research.

---

### 🏦 Example

Interview FinTech managers to understand digital wallet security policies.

---

### ✅ Advantages

* Focused insights
* High relevance
* Useful for qualitative research

---

### ❌ Limitations

* Researcher bias
* Not statistically generalizable

---

## **4.3 Quota Sampling**

---

### 🔎 Definition

Researcher selects participants to match specific population proportions.

---

### 🧠 Simple Explanation

> Ensure sample reflects certain characteristics.

---

### 🏦 Example

Population:
60% male
40% female

Ensure sample also has 60% male and 40% female.

---

### 📌 Difference from Stratified Sampling

Stratified = Random selection within groups
Quota = Non-random selection within groups

---

### ✅ Advantages

* More representative than convenience sampling
* Ensures subgroup representation

---

### ❌ Limitations

* Selection bias
* No true randomization

---

## **4.4 Snowball Sampling**

---

### 🔎 Definition

Existing participants refer new participants.

Used when population is difficult to access.

---

### 🧠 Simple Explanation

> Participants recruit other participants.

---

### 🏦 Example

Studying cryptocurrency traders:

One trader refers another.

---

### ✅ Advantages

* Access hidden populations
* Useful in niche research

---

### ❌ Limitations

* High bias
* Lack of diversity
* Limited generalizability

---

## **4.5 Advantages and Limitations of Non-Probability Sampling**

---

### ✅ Advantages

* Faster
* Cost-effective
* Practical in early research stages
* Useful for qualitative studies

---

### ❌ Limitations

* Sampling bias
* Cannot estimate sampling error
* Limited statistical inference
* Weak generalizability

---

## 📊 Comparison Table

| Method      | Best Used For           | Main Risk            |
| ----------- | ----------------------- | -------------------- |
| Convenience | Pilot studies           | High bias            |
| Judgmental  | Expert interviews       | Researcher bias      |
| Quota       | Subgroup representation | Non-random selection |
| Snowball    | Hidden populations      | Homogeneity bias     |

---

# 🎯 Key Takeaway

Non-probability sampling is useful in exploratory and qualitative research but has limitations in statistical generalization.

# **5. Measurement in Business Research**

Measurement transforms abstract concepts into measurable variables.

---

## **5.1 Meaning of Measurement**

---

### 🔎 Definition

Measurement is the process of assigning numbers or symbols to characteristics according to specific rules.

---

### 🧠 Simple Explanation

> Measurement converts ideas into numbers.

---

### 📌 Why Measurement is Important

Without measurement:

* Hypotheses cannot be tested
* Statistical analysis is impossible
* Research remains conceptual

---

### 🏦 Example

Concept: Trust

Measurement:

“How strongly do you agree that PaySecure is secure?”

Scale: 1–5

Trust becomes measurable.

---

## **5.2 Operationalization of Variables**

---

### 🔎 What is Operationalization?

Operationalization means defining how a concept will be measured.

---

### 🧠 Example

Concept: Customer Satisfaction

Operational Definition:

Average score on 5 survey questions measuring satisfaction.

---

### 🏦 FinTech Example

Concept: Digital Adoption

Operationalization:

Number of transactions per month.

---

### 📌 Steps in Operationalization

1. Define concept clearly
2. Identify measurable indicators
3. Develop scale
4. Test reliability

---

## **5.3 Assigning Numbers to Concepts**

---

In measurement, abstract ideas are converted into numeric form.

---

### 📊 Example

“I trust PaySecure.”

1 = Strongly Disagree
2 = Disagree
3 = Neutral
4 = Agree
5 = Strongly Agree

Now trust becomes quantitative.

---

### 📌 Important

Numbers represent intensity, not exact physical measurement.

---

## **5.4 Measurement Error**

---

### 🔎 What is Measurement Error?

Measurement error is the difference between:

True value
AND
Observed value

---

### 🧠 Causes of Measurement Error

* Poorly worded questions
* Respondent misunderstanding
* Biased survey design
* Data entry errors
* Instrument unreliability

---

### 🏦 Example

Question:
“Don’t you agree that digital wallets are unsafe?”

Leading question → measurement error.

---

### 📌 Types of Measurement Error

* Random error
* Systematic error

---

### 📌 How to Reduce Measurement Error

* Pre-test questionnaire
* Use validated scales
* Clear wording
* Standardized procedures

---

# 📊 Conceptual Flow

Concept → Operationalization → Measurement → Data → Analysis

---

# 🎯 Final Takeaways

* Non-probability sampling is practical but limited in generalization.
* Measurement converts concepts into numbers.
* Operationalization defines how concepts are measured.
* Poor measurement leads to unreliable results.

# **6. Measurement Scales**

Before analyzing data in Python or performing statistical tests, we must first understand:

> What kind of numbers are we working with?

Not all numbers mean the same thing.

For example:

* “Wallet Type = 1” does NOT mean better than “Wallet Type = 2”
* “10 transactions” is truly double “5 transactions”

This difference comes from **measurement scales**.

There are **four major measurement scales**:

1. Nominal
2. Ordinal
3. Interval
4. Ratio

They move from weakest to strongest.

---

# **6.1 Nominal Scale**

---

## 🔎 Definition

Nominal scale is used for **classification only**.

It assigns labels or names to categories.

There is:

* No ranking
* No order
* No meaningful mathematical operations

---

## 🧠 Simple Way to Think

> Nominal = Name only.

---

## 🏦 Business Examples

### Example 1: Wallet Brand

| User | Wallet Used |
| ---- | ----------- |
| U1   | PaySecure   |
| U2   | JazzCash    |
| U3   | EasyPaisa   |

We can code:

PaySecure = 1
JazzCash = 2
EasyPaisa = 3

But:

1 ≠ better than 2
3 ≠ larger than 1

These are just labels.

---

### Example 2: Customer Type

* Individual
* Corporate
* SME

Again, just categories.

---

## 📊 What Can We Calculate?

Allowed:

* Frequency (How many use PaySecure?)
* Percentage
* Mode (Most common category)
* Cross-tabulation
* Chi-square test

Not allowed:

* Mean
* Median
* Standard deviation

Because there is no numerical meaning.

---

## ❌ Common Mistake

Calculating average of:

Male = 1
Female = 2

Average = 1.6

This is meaningless.

---

# **6.2 Ordinal Scale**

---

## 🔎 Definition

Ordinal scale provides:

* Categories
* A meaningful order
* But unequal distances between categories

---

## 🧠 Simple Way to Think

> Ordinal = Rank or order.

---

## 🏦 Business Example 1: Satisfaction Level

| Response | Meaning           |
| -------- | ----------------- |
| 1        | Very Dissatisfied |
| 2        | Dissatisfied      |
| 3        | Neutral           |
| 4        | Satisfied         |
| 5        | Very Satisfied    |

We know:

5 > 4 > 3 > 2 > 1

But:

Difference between 4 and 5 may not equal difference between 1 and 2.

---

## 🏦 Business Example 2: Credit Rating

* Poor
* Fair
* Good
* Excellent

There is ranking.

But how much better is “Excellent” than “Good”?
We don’t know exact distance.

---

## 📊 What Can We Calculate?

Allowed:

* Median
* Percentiles
* Rank-based correlation
* Non-parametric tests

Often (in practice), researchers treat Likert scales as interval data — but conceptually they are ordinal.

---

## ❌ What We Cannot Fully Justify

Strictly speaking:

Mean is questionable because distances may not be equal.

---

# **6.3 Interval Scale**

---

## 🔎 Definition

Interval scale provides:

* Order
* Equal intervals between values
* BUT no true zero

---

## 🧠 Simple Way to Think

> Interval = Equal spacing, but zero is arbitrary.

---

## 🏦 Example 1: Temperature (Celsius)

0°C does NOT mean “no temperature.”

Difference between:

10°C and 20°C = 10
20°C and 30°C = 10

Equal intervals.

But:

40°C is NOT twice as hot as 20°C.

Because zero is not absolute.

---

## 🏦 Business Example 2: Standardized Satisfaction Score (0–100)

If:

Score 60 → Moderate
Score 80 → High

Difference between 60 and 70 = 10
Difference between 80 and 90 = 10

Equal spacing.

But 80 is not double 40 in real satisfaction sense.

---

## 📊 What Can We Calculate?

Allowed:

* Mean
* Standard deviation
* Correlation
* Regression
* T-test
* ANOVA

Interval data supports parametric statistics.

---

# **6.4 Ratio Scale**

---

## 🔎 Definition

Ratio scale has:

* Order
* Equal intervals
* True zero
* Meaningful ratios

---

## 🧠 Simple Way to Think

> Ratio = Fully measurable.

---

## 🏦 Business Examples

### Example 1: Monthly Transactions

| User | Transactions |
| ---- | ------------ |
| U1   | 10           |
| U2   | 5            |

10 is truly double 5.

0 means no transactions.

---

### Example 2: Income

| User | Income  |
| ---- | ------- |
| A    | 100,000 |
| B    | 50,000  |

User A earns twice as much as User B.

Zero income = no income.

---

### Example 3: Transaction Amount

0 rupees = no transaction.

This is a true zero.

---

## 📊 What Can We Calculate?

Everything:

* Mean
* Variance
* Regression
* Forecasting
* Machine learning
* Ratios
* Growth rates

Ratio scale is the most powerful for analytics.

---

# **6.5 Full Comparison with Clear Logic**

| Feature           | Nominal | Ordinal | Interval | Ratio |
| ----------------- | ------- | ------- | -------- | ----- |
| Labels            | Yes     | Yes     | Yes      | Yes   |
| Order             | No      | Yes     | Yes      | Yes   |
| Equal spacing     | No      | No      | Yes      | Yes   |
| True zero         | No      | No      | No       | Yes   |
| Mean allowed      | No      | Limited | Yes      | Yes   |
| Ratios meaningful | No      | No      | No       | Yes   |

---

# **6.6 Practical FinTech Research Example**

Suppose you collect:

1. Wallet Brand → Nominal
2. Trust Level (1–5 Likert) → Ordinal
3. Satisfaction Score (0–100) → Interval
4. Monthly Transaction Count → Ratio

If you want to run regression:

Dependent variable should ideally be Ratio or Interval.

Nominal variables must be converted into dummy variables.

---

# **6.7 Why Measurement Scale Matters in Python**

Before running models:

* Identify scale
* Encode nominal data
* Avoid calculating mean of categories
* Choose correct statistical test

Wrong scale → wrong model → invalid research.

---

# 🎯 Final Clear Understanding

* Nominal → Category only
* Ordinal → Rank only
* Interval → Equal distance, no true zero
* Ratio → Equal distance + true zero

Strongest scale = Ratio
Weakest scale = Nominal
