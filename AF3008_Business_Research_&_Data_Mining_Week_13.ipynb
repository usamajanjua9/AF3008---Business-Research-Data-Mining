# 🔹 **1. Introduction to Clustering**

---

# 📌 **1.1 What is Clustering?**

## ✅ **Definition**

👉 **Clustering = Grouping similar data points together**

---

## 💡 In simple words:

> Clustering means putting similar things into the same group **without knowing the groups beforehand**

---

## 🎯 Example:

Suppose you have customers:

* Some spend a lot 💰
* Some spend average
* Some spend very little

👉 Clustering will automatically group them into:

* High spenders
* Medium spenders
* Low spenders

---

## 🧠 Key Idea:

* No labels are given
* Model finds groups **on its own**

---

## 🔍 Important:

👉 In clustering:

* We don’t tell the model:

  * “This is high spender”
* The model discovers it automatically

---

---

# 📌 **1.2 Why Clustering is Important**

Clustering is very useful when we **don’t know categories in advance**.

---

## ✅ 1. Discover Hidden Patterns

👉 Data may contain patterns that are not visible

Example:

* Customers behave differently
* But we don’t know how many types exist

---

## ✅ 2. Understand Data Better

Clustering helps answer:

* How many types of customers exist?
* Are there similar groups?

---

## ✅ 3. Better Decision Making

Businesses can:

* target specific groups
* design better strategies

---

## ✅ 4. Personalization

Companies can:

* show personalized ads
* recommend products

---

## 🧠 Simple Summary:

> Clustering helps us **explore and understand data without prior knowledge**

---

---

# 📌 **1.3 Clustering vs Classification**

⚠️ This is very important, students often confuse these.

---

## 🔥 **Main Difference**

| Feature          | Clustering      | Classification    |

| Learning Type    | Unsupervised    | Supervised        |
| Labels Available | ❌ No            | ✅ Yes             |
| Output           | Groups          | Categories        |
| Example          | Customer groups | Fraud / Not Fraud |

---

---

## 🎯 Example Comparison

### 👉 Classification:

We already know categories:

* Spam / Not Spam

---

### 👉 Clustering:

We don’t know categories:

* Model creates groups automatically

---

---

## 🧠 Easy Trick:

👉 **Known labels → Classification**
👉 **Unknown labels → Clustering**

---

---

## 📦 Simple Analogy

### Classroom Example:

  * Pass

👉 Clustering:

👉 In unsupervised learning, there are **no labels**

### 🎯 Example:

| Customer | Spending |

* find patterns

# 📌 **2.2 How Clustering Works**

---

## ✅ **Simple Explanation**

👉 Clustering groups data points based on **similarity**

---

## 🎯 Example:

Suppose we have points:

```text
(2,2), (3,3), (8,8), (9,9)
```

👉 Clustering groups them as:

* Group 1: (2,2), (3,3)
* Group 2: (8,8), (9,9)

---

## 🧠 Why?

Because:

* points close together → same group
* far apart → different group

---

---

## 🔄 **Basic Steps**

1. Start with data
2. Measure similarity
3. Group similar points
4. Repeat until stable

---

---

## 🎯 Real-Life Example

Customers:

| Customer | Spending |
| -------- | -------- |
| A        | 1000     |
| B        | 1200     |
| C        | 9000     |

👉 Clustering:

* Group 1 → Low spenders
* Group 2 → High spenders

---

---

# 📌 **2.3 Similarity and Distance Concept**

---

## ✅ **Definition**

👉 Clustering works by measuring how **close or far** data points are

---

## 🧠 Key Idea:

> Closer points → more similar
> Far points → less similar

---

---

## 📏 **Distance Concept**

We use **distance** to measure similarity.

---

## 🎯 Example:

Points:

* A = (2,2)
* B = (3,3)

Distance is small → similar

---

---

## 🔥 **Euclidean Distance (Most Common)**

---

## 🎯 Example Calculation:

Points:

* A(2,2)
* B(4,4)


$$
d = \sqrt{(4-2)^2 + (4-2)^2}
$$



$$
d = \sqrt{4 + 4} = \sqrt{8}
$$


---

## 🧠 Interpretation:

* Small distance → same cluster
* Large distance → different cluster

---

---

## 📌 Other Distance Types (Basic Mention)

* Manhattan Distance
* Cosine Similarity


---

---

# 📌 **2.4 Types of Clustering (Hard vs Soft Clustering)**

---

## ✅ **1. Hard Clustering**

👉 Each data point belongs to **only ONE cluster**

---

### 🎯 Example:

Customer A:

* High spender

👉 Cannot belong to:

* Medium spender

---

## 🧠 Key Idea:

> One point → One group only

---

---

## ✅ **2. Soft Clustering**

👉 A data point can belong to **multiple clusters**

---

### 🎯 Example:

Customer A:

* 70% high spender
* 30% medium spender

---

## 🧠 Key Idea:

> Partial membership

---

---

## 🔥 **Comparison**

| Feature    | Hard Clustering | Soft Clustering   |
| ---------- | --------------- | ----------------- |
| Membership | One cluster     | Multiple clusters |
| Example    | K-Means         | Fuzzy Clustering  |

# 🔹 **3. K-Means Clustering (Core Algorithm)**

---

# 📌 **3.1 What is K-Means?**

## ✅ **Definition**

👉 **K-Means is a clustering algorithm that groups data into K clusters based on similarity**

---

## 💡 In simple words:

> K-Means divides data into **K groups**, where each group contains similar data points

---

## 🎯 Example:

Suppose you have customers based on spending:

* Low spenders
* Medium spenders
* High spenders

👉 If we set:

```text
K = 3
```

Then K-Means will create **3 clusters automatically**

---

## 🧠 Key Idea:

* K = number of clusters
* Means = average (centroid)

---

---

# 📌 **3.2 Step-by-Step Working of K-Means**

---

## 🎯 **We will use a simple example**

Data points:

```text
(2,2), (3,3), (8,8), (9,9)
```

👉 Goal: Divide into **K = 2 clusters**

---

---

# 🔄 **Step 1: Choose K**

```text
K = 2
```

👉 We want 2 groups

---

---

# 🔄 **Step 2: Initialize Centroids**

👉 Randomly pick 2 points as centroids

Assume:

* C1 = (2,2)
* C2 = (8,8)

---

---

# 🔄 **Step 3: Assign Points to Nearest Centroid**

👉 Use distance formula:

---

## 🎯 Calculate distances

## 🎯 Key Formula

Distance:


$$
d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}
$$
check,
### Point (2,2)

Distance to C1:


$$
\sqrt{(2-2)^2 + (2-2)^2} = 0
$$


Distance to C2:


$$
\sqrt{(8-2)^2 + (8-2)^2} = \sqrt{36+36} = \sqrt{72}
$$


👉 Closer to C1

---

---

### Point (3,3)

Distance to C1:


$$
\sqrt{(3-2)^2 + (3-2)^2} = \sqrt{2}
$$


Distance to C2:


$$
\sqrt{(8-3)^2 + (8-3)^2} = \sqrt{50}
$$


👉 Closer to C1

---

---

### Point (8,8)

Distance to C2 = 0 → assign to C2

---

---

### Point (9,9)

Distance to C2:


$$
\sqrt{(9-8)^2 + (9-8)^2} = \sqrt{2}
$$


👉 Closer to C2

---

---

## ✅ After Step 3:

Cluster 1:

* (2,2), (3,3)

Cluster 2:

* (8,8), (9,9)

---
---

# 🔹 Step 4: Update the Centroids

Now we calculate the new center of each cluster.

## Formula for New Centroid

$$
Centroid = \left(\frac{\sum x}{n}, \frac{\sum y}{n}\right)
$$

Where:

* (\sum x) = sum of x-values
* (\sum y) = sum of y-values
* (n) = number of points in the cluster

---
---

# 🔄 **Step 4: Update Centroids**

👉 Compute mean (average)

---

## Cluster 1:


$$
Centroid = \left(\frac{2+3}{2}, \frac{2+3}{2}\right) = (2.5, 2.5)
$$


---

## Cluster 2:


$$
Centroid = \left(\frac{8+9}{2}, \frac{8+9}{2}\right) = (8.5, 8.5)
$$


---

---

# 🔄 **Step 5: Repeat Assignment**

👉 Assign again using new centroids

👉 (Same clusters, no change)

---

---

# 🔄 **Step 6: Stop Condition**

👉 Algorithm stops when:

* centroids don’t change
* clusters remain same

---

---

# 🔥 **Final Result**

Cluster 1:

* (2,2), (3,3)

Cluster 2:

* (8,8), (9,9)

---

---

# 📌 **3.3 Choosing the Value of K**

---

## ❓ Problem

👉 How many clusters should we choose?

---

## 🎯 Example:

K = 2 → 2 groups
K = 3 → 3 groups

---

---

## 🔥 **Elbow Method (Basic Idea)**

👉 We test different K values:

* K = 1
* K = 2
* K = 3

---

👉 Plot:

* X-axis → K
* Y-axis → Error

---

👉 Choose K where graph bends like an “elbow”

---

## 🧠 Simple Idea:

> Too small K → poor grouping
> Too large K → over grouping

---

---

# 📌 **3.4 Advantages of K-Means**

---

## ✅ 1. Easy to Understand

* Simple steps
* Easy math

---

## ✅ 2. Fast Algorithm

* Works quickly on large data

---

## ✅ 3. Works Well for Clear Clusters

* Good when groups are well separated

---

## ✅ 4. Widely Used

* Business
* Marketing
* FinTech

---

---

# 📌 **3.5 Limitations of K-Means**

---

## ❌ 1. Need to Choose K

👉 Must decide number of clusters beforehand

---

## ❌ 2. Sensitive to Initial Centroids

👉 Different starting points → different results

---

## ❌ 3. Not Good for Complex Shapes

👉 Works best for round clusters

---

## ❌ 4. Affected by Outliers

👉 Extreme values can distort clusters

---

## ❌ 5. Assumes Equal Cluster Size

👉 Not always true in real data

---

---

# 🔥 **Final Summary**

👉 K-Means works like:

1. Choose K
2. Pick centroids
3. Assign points
4. Update centroids
5. Repeat

# 📘 K-Means Clustering: Step-by-Step Manual Example

## 🎯 Goal

We will cluster 6 data points into 2 groups using **K-Means Clustering**.

Suppose we have customer data based on:

* X-axis = Monthly Spending
* Y-axis = Number of Purchases

---

# 🔹 Dataset

| Customer | Point  |
| -------- | ------ |
| A        | (1, 2) |
| B        | (2, 1) |
| C        | (2, 3) |
| D        | (8, 8) |
| E        | (9, 8) |
| F        | (8, 9) |

We will use:

$$
K = 2
$$

This means we want to create **2 clusters**.

---

# 🔹 Main Distance Formula

K-Means uses distance to decide which point is closer to which centroid.

We use **Euclidean Distance**:

$$
d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}
$$

Where:

* ((x_1, y_1)) = data point
* ((x_2, y_2)) = centroid

---

# 🔹 Step 1: Choose Initial Centroids

Let us randomly choose:

$$
C_1 = (1, 2)
$$

$$
C_2 = (8, 8)
$$

---

# 🔹 Step 2: Calculate Distance of Each Point from Both Centroids

## Point A = (1, 2)

Distance from (C_1 = (1, 2)):

$$
d(A, C_1) = \sqrt{(1-1)^2 + (2-2)^2}
$$

$$
d(A, C_1) = \sqrt{0^2 + 0^2} = 0
$$

Distance from (C_2 = (8, 8)):

$$
d(A, C_2) = \sqrt{(1-8)^2 + (2-8)^2}
$$

$$
d(A, C_2) = \sqrt{(-7)^2 + (-6)^2}
$$

$$
d(A, C_2) = \sqrt{49 + 36} = \sqrt{85} \approx 9.22
$$

So, A belongs to **Cluster 1**.

---

## Point B = (2, 1)

Distance from (C_1 = (1, 2)):

$$
d(B, C_1) = \sqrt{(2-1)^2 + (1-2)^2}
$$

$$
d(B, C_1) = \sqrt{1^2 + (-1)^2}
$$

$$
d(B, C_1) = \sqrt{1 + 1} = \sqrt{2} \approx 1.41
$$

Distance from (C_2 = (8, 8)):

$$
d(B, C_2) = \sqrt{(2-8)^2 + (1-8)^2}
$$

$$
d(B, C_2) = \sqrt{(-6)^2 + (-7)^2}
$$

$$
d(B, C_2) = \sqrt{36 + 49} = \sqrt{85} \approx 9.22
$$

So, B belongs to **Cluster 1**.

---

## Point C = (2, 3)

Distance from (C_1 = (1, 2)):

$$
d(C, C_1) = \sqrt{(2-1)^2 + (3-2)^2}
$$

$$
d(C, C_1) = \sqrt{1^2 + 1^2}
$$

$$
d(C, C_1) = \sqrt{2} \approx 1.41
$$

Distance from (C_2 = (8, 8)):

$$
d(C, C_2) = \sqrt{(2-8)^2 + (3-8)^2}
$$

$$
d(C, C_2) = \sqrt{(-6)^2 + (-5)^2}
$$

$$
d(C, C_2) = \sqrt{36 + 25} = \sqrt{61} \approx 7.81
$$

So, C belongs to **Cluster 1**.

---

## Point D = (8, 8)

Distance from (C_1 = (1, 2)):

$$
d(D, C_1) = \sqrt{(8-1)^2 + (8-2)^2}
$$

$$
d(D, C_1) = \sqrt{7^2 + 6^2}
$$

$$
d(D, C_1) = \sqrt{49 + 36} = \sqrt{85} \approx 9.22
$$

Distance from (C_2 = (8, 8)):

$$
d(D, C_2) = \sqrt{(8-8)^2 + (8-8)^2}
$$

$$
d(D, C_2) = 0
$$

So, D belongs to **Cluster 2**.

---

## Point E = (9, 8)

Distance from (C_1 = (1, 2)):

$$
d(E, C_1) = \sqrt{(9-1)^2 + (8-2)^2}
$$

$$
d(E, C_1) = \sqrt{8^2 + 6^2}
$$

$$

Distance from (C_2 = (8, 8)):

$$
d(E, C_2) = \sqrt{(9-8)^2 + (8-8)^2}
$$

$$
d(E, C_2) = \sqrt{1^2 + 0^2}

$$
d(E, C_2) = 1
$$

So, E belongs to **Cluster 2**.

---

## Point F = (8, 9)

Distance from (C_1 = (1, 2)):

$$
d(F, C_1) = \sqrt{(8-1)^2 + (9-2)^2}
$$

$$
d(F, C_1) = \sqrt{7^2 + 7^2}

$$

$$
d(F, C_2) = \sqrt{(-0.33)^2 + 0.67^2}
$$

$$
d(F, C_2) = \sqrt{0.11 + 0.45} = \sqrt{0.56} \approx 0.75
$$

F remains in **Cluster 2**.

---

# 🔹 Step 6: Second Cluster Assignment

| Point   | Distance from New C1 | Distance from New C2 | Assigned Cluster |
| ------- | -------------------: | -------------------: | ---------------- |
| A (1,2) |                 0.67 |                 9.68 | Cluster 1        |
| B (2,1) |                 1.05 |                 9.68 | Cluster 1        |
| C (2,3) |                 1.05 |                 8.28 | Cluster 1        |
| D (8,8) |                 8.72 |                 0.47 | Cluster 2        |
| E (9,8) |                 9.47 |                 0.75 | Cluster 2        |
| F (8,9) |                 9.44 |                 0.75 | Cluster 2        |

The cluster assignment did not change.

So K-Means stops.

---

# 🔹 Final Clusters

## Cluster 1

$$
Cluster\ 1 = {(1,2), (2,1), (2,3)}
$$

This may represent:

> Low spending, low purchase customers

---

## Cluster 2

$$
Cluster\ 2 = {(8,8), (9,8), (8,9)}
$$

This may represent:

> High spending, high purchase customers

---

# 🔹 Final Centroids

$$
C_1 = (1.67, 2)
$$

$$
C_2 = (8.33, 8.33)
$$

---

# 🔹 Complete K-Means Steps Summary

| Step   | Explanation                                     |
| ------ | ----------------------------------------------- |
| Step 1 | Choose number of clusters (K)                   |
| Step 2 | Select initial centroids                        |
| Step 3 | Calculate distance of each point from centroids |
| Step 4 | Assign each point to nearest centroid           |
| Step 5 | Recalculate centroid of each cluster            |
| Step 6 | Repeat until clusters do not change             |

---

# 🔹 Short Final Explanation for Students

K-Means starts with random cluster centers. Then it repeatedly assigns each data point to the nearest center and updates the center using the average of points in that cluster. The process stops when the cluster assignments no longer change.

In this example, after one update, the clusters remained the same, so the algorithm stopped.


````python
# ============================================================
# K-MEANS CLUSTERING - STEP-BY-STEP BEGINNER FRIENDLY CODE
# Example: 6 Customer Data Points
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# ------------------------------------------------------------
# 1. Create Dataset
# ------------------------------------------------------------

data = {
    "Customer": ["A", "B", "C", "D", "E", "F"],
    "Monthly_Spending": [1, 2, 2, 8, 9, 8],
    "Number_of_Purchases": [2, 1, 3, 8, 8, 9]
}

df = pd.DataFrame(data)

print("=" * 70)
print("STEP 1: ORIGINAL DATASET")
print("=" * 70)
print(df)

# Features for clustering
X = df[["Monthly_Spending", "Number_of_Purchases"]].values

# ------------------------------------------------------------
# 2. Visualize Original Points
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
plt.scatter(df["Monthly_Spending"], df["Number_of_Purchases"], s=120)

for i, row in df.iterrows():
    plt.text(row["Monthly_Spending"] + 0.1, row["Number_of_Purchases"] + 0.1, row["Customer"], fontsize=12)

plt.title("Original Customer Data Points")
plt.xlabel("Monthly Spending")
plt.ylabel("Number of Purchases")
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 3. Define K
# ------------------------------------------------------------

K = 2

print("\n" + "=" * 70)
print("STEP 2: SELECT NUMBER OF CLUSTERS")
print("=" * 70)
print(f"K = {K}")
print("This means we want to create 2 customer groups.")

# ------------------------------------------------------------
# 4. Manual Initial Centroids
# ------------------------------------------------------------

# Same initial centroids used in manual explanation
centroids = np.array([
    [1, 2],   # C1
    [8, 8]    # C2
], dtype=float)

print("\n" + "=" * 70)
print("STEP 3: INITIAL CENTROIDS")
print("=" * 70)
print(f"C1 = {centroids[0]}")
print(f"C2 = {centroids[1]}")

# ------------------------------------------------------------
# 5. Euclidean Distance Function
# ------------------------------------------------------------

def euclidean_distance(point, centroid):
    return np.sqrt(np.sum((point - centroid) ** 2))

# ------------------------------------------------------------
# 6. First Iteration: Calculate Distances
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 4: FIRST ITERATION - DISTANCE CALCULATION")
print("=" * 70)

distance_records = []
cluster_assignments = []

for i, point in enumerate(X):
    customer = df.loc[i, "Customer"]

    d1 = euclidean_distance(point, centroids[0])
    d2 = euclidean_distance(point, centroids[1])

    if d1 < d2:
        cluster = "Cluster 1"
    else:
        cluster = "Cluster 2"

    cluster_assignments.append(cluster)

    distance_records.append({
        "Customer": customer,
        "Point": tuple(point),
        "Distance from C1": round(d1, 2),
        "Distance from C2": round(d2, 2),
        "Assigned Cluster": cluster
    })

distance_df = pd.DataFrame(distance_records)
print(distance_df)

df["Iteration_1_Cluster"] = cluster_assignments

# ------------------------------------------------------------
# 7. Update Centroids
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 5: UPDATE CENTROIDS")
print("=" * 70)

new_centroids = []

for cluster_name in ["Cluster 1", "Cluster 2"]:
    cluster_points = df[df["Iteration_1_Cluster"] == cluster_name][["Monthly_Spending", "Number_of_Purchases"]].values

    new_centroid = cluster_points.mean(axis=0)
    new_centroids.append(new_centroid)

    print(f"\n{cluster_name} points:")
    print(cluster_points)
    print(f"New centroid = mean of points = {new_centroid}")

new_centroids = np.array(new_centroids)

print("\nUpdated Centroids:")
print(f"New C1 = {new_centroids[0]}")
print(f"New C2 = {new_centroids[1]}")

# ------------------------------------------------------------
# 8. Visualize After First Assignment
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

for cluster_name in ["Cluster 1", "Cluster 2"]:
    cluster_data = df[df["Iteration_1_Cluster"] == cluster_name]
    plt.scatter(
        cluster_data["Monthly_Spending"],
        cluster_data["Number_of_Purchases"],
        s=120,
        label=cluster_name
    )

for i, row in df.iterrows():
    plt.text(row["Monthly_Spending"] + 0.1, row["Number_of_Purchases"] + 0.1, row["Customer"], fontsize=12)

plt.scatter(new_centroids[:, 0], new_centroids[:, 1], s=250, marker="X", label="New Centroids")

plt.title("Clusters After First Iteration")
plt.xlabel("Monthly Spending")
plt.ylabel("Number of Purchases")
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 9. Second Iteration
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 6: SECOND ITERATION - USING UPDATED CENTROIDS")
print("=" * 70)

distance_records_2 = []
cluster_assignments_2 = []

for i, point in enumerate(X):
    customer = df.loc[i, "Customer"]

    d1 = euclidean_distance(point, new_centroids[0])
    d2 = euclidean_distance(point, new_centroids[1])

    if d1 < d2:
        cluster = "Cluster 1"
    else:
        cluster = "Cluster 2"

    cluster_assignments_2.append(cluster)

    distance_records_2.append({
        "Customer": customer,
        "Point": tuple(point),
        "Distance from New C1": round(d1, 2),
        "Distance from New C2": round(d2, 2),
        "Assigned Cluster": cluster
    })

distance_df_2 = pd.DataFrame(distance_records_2)
print(distance_df_2)

df["Iteration_2_Cluster"] = cluster_assignments_2

# ------------------------------------------------------------
# 10. Check Stopping Condition
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 7: STOPPING CONDITION")
print("=" * 70)

if list(df["Iteration_1_Cluster"]) == list(df["Iteration_2_Cluster"]):
    print("Cluster assignments did NOT change.")
    print("K-Means stops here.")
else:
    print("Cluster assignments changed.")
    print("K-Means should continue.")

print("\nFinal Dataset with Clusters:")
print(df)

# ------------------------------------------------------------
# 11. Final Visualization
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

for cluster_name in ["Cluster 1", "Cluster 2"]:
    cluster_data = df[df["Iteration_2_Cluster"] == cluster_name]
    plt.scatter(
        cluster_data["Monthly_Spending"],
        cluster_data["Number_of_Purchases"],
        s=120,
        label=cluster_name
    )

for i, row in df.iterrows():
    plt.text(row["Monthly_Spending"] + 0.1, row["Number_of_Purchases"] + 0.1, row["Customer"], fontsize=12)

plt.scatter(new_centroids[:, 0], new_centroids[:, 1], s=250, marker="X", label="Final Centroids")

plt.title("Final K-Means Clusters")
plt.xlabel("Monthly Spending")
plt.ylabel("Number of Purchases")
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 12. Same Example Using sklearn KMeans
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 8: USING SCIKIT-LEARN KMEANS")
print("=" * 70)

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
df["Sklearn_Cluster"] = kmeans.fit_predict(X)

print(df[["Customer", "Monthly_Spending", "Number_of_Purchases", "Sklearn_Cluster"]])

print("\nSklearn Centroids:")
print(kmeans.cluster_centers_)

# ------------------------------------------------------------
# 13. Sklearn Visualization
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))
plt.scatter(
    df["Monthly_Spending"],
    df["Number_of_Purchases"],
    c=df["Sklearn_Cluster"],
    s=120
)

for i, row in df.iterrows():
    plt.text(row["Monthly_Spending"] + 0.1, row["Number_of_Purchases"] + 0.1, row["Customer"], fontsize=12)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    s=250,
    marker="X",
    label="Sklearn Centroids"
)

plt.title("K-Means Clustering using Scikit-Learn")
plt.xlabel("Monthly Spending")
plt.ylabel("Number of Purchases")
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 14. Final Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)
print("K-Means grouped customers into 2 clusters:")
print("Cluster 1: Low spending and low purchase customers")
print("Cluster 2: High spending and high purchase customers")
print("\nMain steps:")
print("1. Choose K")
print("2. Select initial centroids")
print("3. Calculate distances")
print("4. Assign points to nearest centroid")
print("5. Update centroids")
print("6. Repeat until assignments do not change")
````
